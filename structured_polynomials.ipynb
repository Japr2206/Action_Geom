```python
# =========================================================
# WEEK 0 — Compact scan for self-reciprocal structured polynomials
# Colab / Python only / Plotly dark retro style
# Family:
#   P_{n,m}^sigma(z) = z^(2n) + sigma z^(n+m) + sigma z^(n-m) + 1
# Compact outputs:
#   1) top 10 cases
#   2) best case for each n
#   3) one heat-style scatter
#   4) roots + trig plot for the top case
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 20
UNIT_TOL = 1e-7
SAVE_CSV = True

# Retro palette
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

# -----------------------------
# Helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(text=title or "", x=0.5, xanchor="center",
                   font=dict(size=20, color=NEON_AMBER)),
        xaxis=dict(
            showgrid=True, gridcolor=GRID,
            zeroline=True, zerolinecolor="rgba(255,176,0,0.25)",
            showline=True, linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True, gridcolor=GRID,
            zeroline=True, zerolinecolor="rgba(255,176,0,0.25)",
            showline=True, linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {"selector": "table", "props": [("background-color", BG), ("color", TEXT),
                                            ("border-collapse", "collapse"),
                                            ("font-family", "Courier New, monospace")]},
            {"selector": "th", "props": [("background-color", "#111111"), ("color", NEON_AMBER),
                                         ("border", "1px solid #333333"), ("padding", "6px")]},
            {"selector": "td", "props": [("background-color", "#050505"), ("color", TEXT),
                                         ("border", "1px solid #222222"), ("padding", "6px")]},
            {"selector": "caption", "props": [("color", NEON_GREEN), ("font-size", "16px"),
                                              ("font-weight", "bold"), ("caption-side", "top"),
                                              ("padding", "8px")]},
        ])
    )
    if "U_numeric" in show.columns:
        sty = sty.background_gradient(cmap="viridis", subset=["U_numeric"])
    display(sty)

# -----------------------------
# Polynomial family
# -----------------------------
def poly_coeffs(n, m, sigma):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[2*n - (n + m)] = float(sigma)
    coeffs[2*n - (n - m)] = float(sigma)
    coeffs[-1] = 1.0
    return coeffs

def poly_roots(n, m, sigma):
    return np.roots(poly_coeffs(n, m, sigma))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

def trig_values(n, m, sigma, num=5000):
    t = np.linspace(0, 2*np.pi, num=num)
    y = 2*np.cos(n*t) + 2*sigma*np.cos(m*t)
    return t, y

# -----------------------------
# Scan
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for m in range(1, n):
        for sigma in (-1, 1):
            roots = poly_roots(n, m, sigma)
            U = count_unimodular_roots(roots)
            rows.append({
                "n": n,
                "m": m,
                "sigma": sigma,
                "degree": 2*n,
                "U_numeric": U,
                "all_unimodular_numeric": (U == 2*n),
                "gcd_nm": np.gcd(n, m),
            })

df = pd.DataFrame(rows)

# -----------------------------
# Compact core tables
# -----------------------------
top10 = (
    df.sort_values(by=["U_numeric", "degree", "n", "m", "sigma"],
                   ascending=[False, True, True, True, True])
      .head(10)
      .reset_index(drop=True)
)

best_by_n = (
    df.sort_values(by=["n", "U_numeric", "m", "sigma"],
                   ascending=[True, False, True, True])
      .groupby("n", as_index=False)
      .first()
      .reset_index(drop=True)
)

all_uni = df[df["all_unimodular_numeric"]].sort_values(["n", "m", "sigma"]).reset_index(drop=True)

print(f"Total cases scanned: {len(df)}")
print(f"Max U_numeric found: {int(df['U_numeric'].max())}")
print(f"Cases attaining max: {int((df['U_numeric'] == df['U_numeric'].max()).sum())}")
print(f"Cases with all roots numerically unimodular: {int(df['all_unimodular_numeric'].sum())}")

retro_table(top10, "Top 10 cases by U_numeric", max_rows=10)
retro_table(best_by_n, "Best case for each n", max_rows=25)

if len(all_uni) > 0:
    retro_table(all_uni, "Cases with all roots numerically on the unit circle", max_rows=20)
else:
    print("\nNo cases detected with all roots numerically on the unit circle in this scan.")

# -----------------------------
# Plot 1: compact family map
# -----------------------------
fig1 = px.scatter(
    df,
    x="n",
    y="m",
    color="U_numeric",
    symbol="sigma",
    size="U_numeric",
    hover_data=["degree", "gcd_nm", "all_unimodular_numeric"],
    color_continuous_scale="Viridis",
    title="Compact family map: each point is P_(n,m)^sigma"
)
apply_retro(fig1)
fig1.update_layout(xaxis_title="n", yaxis_title="m")
fig1.show()

# -----------------------------
# Plot 2 and 3: top case only
# -----------------------------
row = top10.iloc[0]
n0, m0, s0 = int(row["n"]), int(row["m"]), int(row["sigma"])
roots0 = poly_roots(n0, m0, s0)

theta = np.linspace(0, 2*np.pi, 800)
fig2 = go.Figure()
fig2.add_trace(go.Scatter(
    x=np.cos(theta), y=np.sin(theta),
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="unit circle"
))
fig2.add_trace(go.Scatter(
    x=np.real(roots0), y=np.imag(roots0),
    mode="markers",
    marker=dict(size=9, color=NEON_CYAN, line=dict(color=NEON_GREEN, width=1)),
    name="roots"
))
apply_retro(fig2, f"Top case roots: n={n0}, m={m0}, sigma={s0}, U={int(row['U_numeric'])}")
fig2.update_layout(xaxis_title="Re", yaxis_title="Im",
                   yaxis_scaleanchor="x", yaxis_scaleratio=1)
fig2.show()

t, y = trig_values(n0, m0, s0, num=5000)
fig3 = go.Figure()
fig3.add_trace(go.Scatter(
    x=t, y=y,
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="2cos(nt)+2sigma cos(mt)"
))
fig3.add_trace(go.Scatter(
    x=[t.min(), t.max()], y=[0, 0],
    mode="lines",
    line=dict(color=NEON_GREEN, width=1, dash="dash"),
    name="y=0"
))
apply_retro(fig3, f"Top case trigonometric profile: n={n0}, m={m0}, sigma={s0}")
fig3.update_layout(xaxis_title="t", yaxis_title="value")
fig3.show()

# -----------------------------
# Save CSV
# -----------------------------
if SAVE_CSV:
    csv_path = "/content/week0_scan_compact.csv"
    df.to_csv(csv_path, index=False)
    print(f"\nSaved full scan to: {csv_path}")
```

```python
# =========================================================
# WEEK 1 START — Nondegenerate sparse reciprocal family
# Colab / Python only / Plotly dark retro style
#
# Family:
#   Q_{n,m}^{alpha,beta}(z) = z^(2n) + alpha z^(n+m) + beta z^n + alpha z^(n-m) + 1
#   alpha, beta in {-1,1}, 1 <= m < n
#
# Outputs:
#   1) top 12 cases by U_numeric
#   2) best case for each n
#   3) distribution of U_numeric
#   4) compact family map
#   5) roots + trig profile for the top case
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 20
UNIT_TOL = 1e-7
SAVE_CSV = True

# Retro palette
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    if "U_numeric" in show.columns:
        sty = sty.background_gradient(cmap="viridis", subset=["U_numeric"])
    display(sty)

# -----------------------------
# Polynomial family
# -----------------------------
def q_coeffs(n, m, alpha, beta):
    """
    Coefficients of
    Q(z) = z^(2n) + alpha z^(n+m) + beta z^n + alpha z^(n-m) + 1
    in descending powers for numpy.roots.
    """
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[2*n - (n + m)] = float(alpha)
    coeffs[2*n - n] = float(beta)
    coeffs[2*n - (n - m)] = float(alpha)
    coeffs[-1] = 1.0
    return coeffs

def q_roots(n, m, alpha, beta):
    return np.roots(q_coeffs(n, m, alpha, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

def trig_values(n, m, alpha, beta, num=6000):
    """
    For z = exp(it):
    e^(-int) Q(e^{it}) = 2 cos(nt) + 2 alpha cos(mt) + beta
    """
    t = np.linspace(0, 2*np.pi, num=num)
    y = 2*np.cos(n*t) + 2*alpha*np.cos(m*t) + beta
    return t, y

# -----------------------------
# Scan
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for m in range(1, n):
        for alpha in (-1, 1):
            for beta in (-1, 1):
                roots = q_roots(n, m, alpha, beta)
                U = count_unimodular_roots(roots)

                rows.append({
                    "n": n,
                    "m": m,
                    "alpha": alpha,
                    "beta": beta,
                    "degree": 2*n,
                    "U_numeric": U,
                    "all_unimodular_numeric": (U == 2*n),
                    "gcd_nm": np.gcd(n, m),
                })

df = pd.DataFrame(rows)

# IMPORTANT FIX:
# create sign_pair BEFORE building summary tables that use it
df["sign_pair"] = df["alpha"].astype(str) + "," + df["beta"].astype(str)

# -----------------------------
# Compact summaries
# -----------------------------
top12 = (
    df.sort_values(
        by=["U_numeric", "degree", "n", "m", "alpha", "beta"],
        ascending=[False, True, True, True, True, True]
    )
    .head(12)
    .reset_index(drop=True)
)

best_by_n = (
    df.sort_values(
        by=["n", "U_numeric", "m", "alpha", "beta"],
        ascending=[True, False, True, True, True]
    )
    .groupby("n", as_index=False)
    .first()
    .reset_index(drop=True)
)

u_distribution = (
    df.groupby("U_numeric", as_index=False)
      .size()
      .rename(columns={"size": "count"})
      .sort_values("U_numeric")
      .reset_index(drop=True)
)

all_uni = (
    df[df["all_unimodular_numeric"]]
    .sort_values(["n", "m", "alpha", "beta"])
    .reset_index(drop=True)
)

print(f"Total cases scanned: {len(df)}")
print(f"Distinct U values found: {sorted(df['U_numeric'].unique().tolist())}")
print(f"Min U_numeric: {int(df['U_numeric'].min())}")
print(f"Max U_numeric: {int(df['U_numeric'].max())}")
print(f"Cases attaining max: {int((df['U_numeric'] == df['U_numeric'].max()).sum())}")
print(f"Cases with all roots numerically unimodular: {int(df['all_unimodular_numeric'].sum())}")

retro_table(top12, "Top 12 cases by U_numeric", max_rows=12)
retro_table(best_by_n, "Best case for each n", max_rows=25)
retro_table(u_distribution, "Distribution of U_numeric", max_rows=30)

if len(all_uni) > 0:
    retro_table(
        all_uni.head(20),
        "First 20 cases with all roots numerically unimodular",
        max_rows=20
    )
else:
    print("\nNo cases detected with all roots numerically on the unit circle in this scan.")

# -----------------------------
# Plot 1: compact family map
# -----------------------------
fig1 = px.scatter(
    df,
    x="n",
    y="m",
    color="U_numeric",
    symbol="sign_pair",
    size="U_numeric",
    hover_data=["degree", "gcd_nm", "alpha", "beta", "all_unimodular_numeric"],
    color_continuous_scale="Viridis",
    title="Compact family map: Q_(n,m)^(alpha,beta)"
)
apply_retro(fig1)
fig1.update_layout(xaxis_title="n", yaxis_title="m")
fig1.show()

# -----------------------------
# Plot 2: best U by n
# -----------------------------
fig2 = px.bar(
    best_by_n,
    x="n",
    y="U_numeric",
    color="sign_pair",
    text="m",
    barmode="group",
    title="Best unimodular-root count for each n"
)
apply_retro(fig2)
fig2.update_traces(textposition="outside")
fig2.update_layout(xaxis_title="n", yaxis_title="Best U(P)")
fig2.show()

# -----------------------------
# Plot 3 and 4: top case only
# -----------------------------
row = top12.iloc[0]
n0 = int(row["n"])
m0 = int(row["m"])
a0 = int(row["alpha"])
b0 = int(row["beta"])
U0 = int(row["U_numeric"])

roots0 = q_roots(n0, m0, a0, b0)

theta = np.linspace(0, 2*np.pi, 800)
fig3 = go.Figure()
fig3.add_trace(go.Scatter(
    x=np.cos(theta),
    y=np.sin(theta),
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="unit circle"
))
fig3.add_trace(go.Scatter(
    x=np.real(roots0),
    y=np.imag(roots0),
    mode="markers",
    marker=dict(size=9, color=NEON_CYAN, line=dict(color=NEON_GREEN, width=1)),
    name="roots"
))
apply_retro(fig3, f"Top case roots: n={n0}, m={m0}, alpha={a0}, beta={b0}, U={U0}")
fig3.update_layout(
    xaxis_title="Re",
    yaxis_title="Im",
    yaxis_scaleanchor="x",
    yaxis_scaleratio=1
)
fig3.show()

t, y = trig_values(n0, m0, a0, b0, num=6000)
fig4 = go.Figure()
fig4.add_trace(go.Scatter(
    x=t,
    y=y,
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="2cos(nt)+2alpha cos(mt)+beta"
))
fig4.add_trace(go.Scatter(
    x=[t.min(), t.max()],
    y=[0, 0],
    mode="lines",
    line=dict(color=NEON_GREEN, width=1, dash="dash"),
    name="y=0"
))
apply_retro(fig4, f"Top case trigonometric profile: n={n0}, m={m0}, alpha={a0}, beta={b0}")
fig4.update_layout(xaxis_title="t", yaxis_title="value")
fig4.show()

# -----------------------------
# Save CSV
# -----------------------------
if SAVE_CSV:
    csv_path = "/content/week1_scan_Q_family.csv"
    df.to_csv(csv_path, index=False)
    print(f"\nSaved full scan to: {csv_path}")
```

```python
# =========================================================
# WEEK 2 START — Primitive subclass gcd(n,m)=1
# Colab / Python only / Plotly dark retro style
#
# Family:
#   Q_{n,m}^{alpha,beta}(z) = z^(2n) + alpha z^(n+m) + beta z^n + alpha z^(n-m) + 1
#   alpha, beta in {-1,1}, 1 <= m < n, gcd(n,m)=1
#
# Compact outputs:
#   1) top 12 primitive cases by U_numeric
#   2) best primitive case for each n
#   3) distribution of U_numeric in primitive class
#   4) compact family map
#   5) roots + trig profile for the top primitive case
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go
from math import gcd

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 20
UNIT_TOL = 1e-7
SAVE_CSV = True

# Retro palette
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    if "U_numeric" in show.columns:
        sty = sty.background_gradient(cmap="viridis", subset=["U_numeric"])
    display(sty)

# -----------------------------
# Polynomial family
# -----------------------------
def q_coeffs(n, m, alpha, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[2*n - (n + m)] = float(alpha)
    coeffs[2*n - n] = float(beta)
    coeffs[2*n - (n - m)] = float(alpha)
    coeffs[-1] = 1.0
    return coeffs

def q_roots(n, m, alpha, beta):
    return np.roots(q_coeffs(n, m, alpha, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

def trig_values(n, m, alpha, beta, num=6000):
    t = np.linspace(0, 2*np.pi, num=num)
    y = 2*np.cos(n*t) + 2*alpha*np.cos(m*t) + beta
    return t, y

# -----------------------------
# Scan primitive subclass only
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for m in range(1, n):
        if gcd(n, m) != 1:
            continue
        for alpha in (-1, 1):
            for beta in (-1, 1):
                roots = q_roots(n, m, alpha, beta)
                U = count_unimodular_roots(roots)

                rows.append({
                    "n": n,
                    "m": m,
                    "alpha": alpha,
                    "beta": beta,
                    "degree": 2*n,
                    "U_numeric": U,
                    "all_unimodular_numeric": (U == 2*n),
                    "gcd_nm": gcd(n, m),
                })

df = pd.DataFrame(rows)

if len(df) == 0:
    raise ValueError("No primitive cases found. Check N_MAX.")

df["sign_pair"] = df["alpha"].astype(str) + "," + df["beta"].astype(str)

# -----------------------------
# Compact summaries
# -----------------------------
top12 = (
    df.sort_values(
        by=["U_numeric", "degree", "n", "m", "alpha", "beta"],
        ascending=[False, True, True, True, True, True]
    )
    .head(12)
    .reset_index(drop=True)
)

best_by_n = (
    df.sort_values(
        by=["n", "U_numeric", "m", "alpha", "beta"],
        ascending=[True, False, True, True, True]
    )
    .groupby("n", as_index=False)
    .first()
    .reset_index(drop=True)
)

u_distribution = (
    df.groupby("U_numeric", as_index=False)
      .size()
      .rename(columns={"size": "count"})
      .sort_values("U_numeric")
      .reset_index(drop=True)
)

all_uni = (
    df[df["all_unimodular_numeric"]]
    .sort_values(["n", "m", "alpha", "beta"])
    .reset_index(drop=True)
)

print(f"Total primitive cases scanned: {len(df)}")
print(f"Distinct U values found (primitive): {sorted(df['U_numeric'].unique().tolist())}")
print(f"Min U_numeric (primitive): {int(df['U_numeric'].min())}")
print(f"Max U_numeric (primitive): {int(df['U_numeric'].max())}")
print(f"Primitive cases attaining max: {int((df['U_numeric'] == df['U_numeric'].max()).sum())}")
print(f"Primitive cases with all roots numerically unimodular: {int(df['all_unimodular_numeric'].sum())}")

retro_table(top12, "Top 12 primitive cases by U_numeric", max_rows=12)
retro_table(best_by_n, "Best primitive case for each n", max_rows=25)
retro_table(u_distribution, "Distribution of U_numeric in primitive class", max_rows=30)

if len(all_uni) > 0:
    retro_table(
        all_uni.head(20),
        "First 20 primitive cases with all roots numerically unimodular",
        max_rows=20
    )
else:
    print("\nNo primitive cases detected with all roots numerically on the unit circle in this scan.")

# -----------------------------
# Plot 1: compact primitive family map
# -----------------------------
fig1 = px.scatter(
    df,
    x="n",
    y="m",
    color="U_numeric",
    symbol="sign_pair",
    size="U_numeric",
    hover_data=["degree", "gcd_nm", "alpha", "beta", "all_unimodular_numeric"],
    color_continuous_scale="Viridis",
    title="Primitive family map: Q_(n,m)^(alpha,beta), gcd(n,m)=1"
)
apply_retro(fig1)
fig1.update_layout(xaxis_title="n", yaxis_title="m")
fig1.show()

# -----------------------------
# Plot 2: best primitive U by n
# -----------------------------
fig2 = px.bar(
    best_by_n,
    x="n",
    y="U_numeric",
    color="sign_pair",
    text="m",
    barmode="group",
    title="Best primitive unimodular-root count for each n"
)
apply_retro(fig2)
fig2.update_traces(textposition="outside")
fig2.update_layout(xaxis_title="n", yaxis_title="Best primitive U(P)")
fig2.show()

# -----------------------------
# Plot 3 and 4: top primitive case only
# -----------------------------
row = top12.iloc[0]
n0 = int(row["n"])
m0 = int(row["m"])
a0 = int(row["alpha"])
b0 = int(row["beta"])
U0 = int(row["U_numeric"])

roots0 = q_roots(n0, m0, a0, b0)

theta = np.linspace(0, 2*np.pi, 800)
fig3 = go.Figure()
fig3.add_trace(go.Scatter(
    x=np.cos(theta),
    y=np.sin(theta),
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="unit circle"
))
fig3.add_trace(go.Scatter(
    x=np.real(roots0),
    y=np.imag(roots0),
    mode="markers",
    marker=dict(size=9, color=NEON_CYAN, line=dict(color=NEON_GREEN, width=1)),
    name="roots"
))
apply_retro(fig3, f"Top primitive case roots: n={n0}, m={m0}, alpha={a0}, beta={b0}, U={U0}")
fig3.update_layout(
    xaxis_title="Re",
    yaxis_title="Im",
    yaxis_scaleanchor="x",
    yaxis_scaleratio=1
)
fig3.show()

t, y = trig_values(n0, m0, a0, b0, num=6000)
fig4 = go.Figure()
fig4.add_trace(go.Scatter(
    x=t,
    y=y,
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="2cos(nt)+2alpha cos(mt)+beta"
))
fig4.add_trace(go.Scatter(
    x=[t.min(), t.max()],
    y=[0, 0],
    mode="lines",
    line=dict(color=NEON_GREEN, width=1, dash="dash"),
    name="y=0"
))
apply_retro(fig4, f"Top primitive trigonometric profile: n={n0}, m={m0}, alpha={a0}, beta={b0}")
fig4.update_layout(xaxis_title="t", yaxis_title="value")
fig4.show()

# -----------------------------
# Save CSV
# -----------------------------
if SAVE_CSV:
    csv_path = "/content/week2_scan_Q_primitive_family.csv"
    df.to_csv(csv_path, index=False)
    print(f"\nSaved primitive scan to: {csv_path}")
```

```python
# =========================================================
# WEEK 3 — Edge regime scan for the primitive family
# Colab / Python only / Plotly dark retro style
#
# Family:
#   Q_{n,m}^{alpha,beta}(z) = z^(2n) + alpha z^(n+m) + beta z^n + alpha z^(n-m) + 1
#
# Edge regime:
#   m = n - k, with small k and gcd(n,k)=1
#
# Goals:
#   1) test whether primitive maximizers live in the edge regime
#   2) test the candidate conjecture
#      M(n) ?= 2n - 2*floor((n+1)/6)
#   3) fix the Plotly legend/colorbar overlap
# =========================================================

import numpy as np
import pandas as pd
from math import gcd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 60          # scan n = 2,...,N_MAX
K_MAX = 6           # edge regime m = n-k with 1 <= k <= K_MAX
UNIT_TOL = 1e-7
SAVE_CSV = True

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

# -----------------------------
# Plot styling
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=180, t=80, b=60),
    )
    return fig

def fix_colorbar(fig):
    fig.update_layout(
        coloraxis_colorbar=dict(
            title="U_numeric",
            x=1.16,
            y=0.5,
            len=0.78,
            thickness=16
        )
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    if "U_numeric" in show.columns:
        sty = sty.background_gradient(cmap="viridis", subset=["U_numeric"])
    display(sty)

# -----------------------------
# Polynomial family
# -----------------------------
def q_coeffs(n, m, alpha, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[2*n - (n + m)] = float(alpha)
    coeffs[2*n - n] = float(beta)
    coeffs[2*n - (n - m)] = float(alpha)
    coeffs[-1] = 1.0
    return coeffs

def q_roots(n, m, alpha, beta):
    return np.roots(q_coeffs(n, m, alpha, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

# -----------------------------
# Candidate conjecture
# -----------------------------
def M_conjectured(n):
    return 2*n - 2*((n + 1) // 6)

# -----------------------------
# Edge scan
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for k in range(1, min(K_MAX, n-1) + 1):
        if gcd(n, k) != 1:
            continue
        m = n - k
        for alpha in (-1, 1):
            for beta in (-1, 1):
                roots = q_roots(n, m, alpha, beta)
                U = count_unimodular_roots(roots)
                rows.append({
                    "n": n,
                    "k": k,
                    "m": m,
                    "alpha": alpha,
                    "beta": beta,
                    "degree": 2*n,
                    "U_numeric": U,
                    "deficit": 2*n - U,
                    "all_unimodular_numeric": (U == 2*n),
                    "sign_pair": f"{alpha},{beta}",
                    "M_conj": M_conjectured(n),
                })

df = pd.DataFrame(rows)

if len(df) == 0:
    raise ValueError("No edge-regime cases found. Increase N_MAX or K_MAX.")

# -----------------------------
# Summaries
# -----------------------------
top15 = (
    df.sort_values(
        by=["U_numeric", "n", "k", "alpha", "beta"],
        ascending=[False, True, True, True, True]
    )
    .head(15)
    .reset_index(drop=True)
)

best_by_n = (
    df.sort_values(
        by=["n", "U_numeric", "k", "alpha", "beta"],
        ascending=[True, False, True, True, True]
    )
    .groupby("n", as_index=False)
    .first()
    .reset_index(drop=True)
)

best_by_n["matches_conjecture"] = (best_by_n["U_numeric"] == best_by_n["M_conj"])

u_distribution = (
    df.groupby("U_numeric", as_index=False)
      .size()
      .rename(columns={"size": "count"})
      .sort_values("U_numeric")
      .reset_index(drop=True)
)

print(f"Total edge-regime cases scanned: {len(df)}")
print(f"Distinct U values found (edge regime): {sorted(df['U_numeric'].unique().tolist())}")
print(f"Max U in edge regime: {int(df['U_numeric'].max())}")
print(f"Cases attaining edge max: {int((df['U_numeric'] == df['U_numeric'].max()).sum())}")
print(f"All-unimodular edge cases: {int(df['all_unimodular_numeric'].sum())}")

retro_table(top15, "Top 15 edge-regime cases by U_numeric", max_rows=15)
retro_table(best_by_n, "Best edge-regime case for each n", max_rows=60)
retro_table(u_distribution, "Distribution of U_numeric in edge regime", max_rows=30)

print("\nConjecture check summary:")
print(best_by_n[["n", "k", "m", "alpha", "beta", "U_numeric", "M_conj", "matches_conjecture"]].to_string(index=False))

# -----------------------------
# Plot 1: edge regime map
# -----------------------------
fig1 = px.scatter(
    df,
    x="n",
    y="k",
    color="U_numeric",
    symbol="sign_pair",
    size="U_numeric",
    hover_data=["m", "alpha", "beta", "deficit", "all_unimodular_numeric"],
    color_continuous_scale="Viridis",
    title="Week 3 edge regime map: m = n - k, gcd(n,k)=1"
)
apply_retro(fig1)
fix_colorbar(fig1)
fig1.update_layout(xaxis_title="n", yaxis_title="k")
fig1.show()

# -----------------------------
# Plot 2: best U vs conjecture
# -----------------------------
fig2 = go.Figure()

fig2.add_trace(go.Scatter(
    x=best_by_n["n"],
    y=best_by_n["U_numeric"],
    mode="lines+markers",
    name="best edge-regime U",
    line=dict(width=3, color=NEON_CYAN),
    marker=dict(size=7, color=NEON_CYAN)
))

fig2.add_trace(go.Scatter(
    x=best_by_n["n"],
    y=best_by_n["M_conj"],
    mode="lines+markers",
    name="candidate conjecture",
    line=dict(width=2, dash="dash", color=NEON_AMBER),
    marker=dict(size=6, color=NEON_AMBER)
))

apply_retro(fig2, "Best edge-regime U versus candidate conjecture")
fig2.update_layout(xaxis_title="n", yaxis_title="U")
fig2.show()

# -----------------------------
# Plot 3: best k by n
# -----------------------------
fig3 = px.bar(
    best_by_n,
    x="n",
    y="k",
    color="sign_pair",
    text="U_numeric",
    barmode="group",
    title="Which k wins?  (m = n-k)"
)
apply_retro(fig3)
fig3.update_traces(textposition="outside")
fig3.update_layout(xaxis_title="n", yaxis_title="best k")
fig3.show()

# -----------------------------
# Save CSV
# -----------------------------
if SAVE_CSV:
    csv_path = "/content/week3_edge_regime_scan.csv"
    df.to_csv(csv_path, index=False)
    print(f"\nSaved edge-regime scan to: {csv_path}")
```

```python
# =========================================================
# WEEK 3B — Focused scan on Q_{n,n-1}^{-1,beta}
# Colab / Python only / Plotly dark retro style
#
# Family:
#   Q_n^beta(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#   beta in {-1,1}
#
# Goals:
#   1) decide which beta wins for each n
#   2) record U_numeric for both beta values
#   3) look for modular patterns in the winning beta
#   4) produce compact tables and clean plots
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 200
UNIT_TOL = 1e-7
SAVE_CSV = True

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=180, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    if "U_numeric" in show.columns:
        sty = sty.background_gradient(cmap="viridis", subset=["U_numeric"])
    display(sty)

# -----------------------------
# Polynomial family
# -----------------------------
def q_edge_coeffs(n, beta):
    """
    Q_n^beta(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
    """
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0          # z^(2n)
    coeffs[1] = -1.0         # z^(2n-1)
    coeffs[n] = float(beta)  # z^n
    coeffs[-2] = -1.0        # z
    coeffs[-1] = 1.0         # 1
    return coeffs

def q_edge_roots(n, beta):
    return np.roots(q_edge_coeffs(n, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

def trig_values(n, beta, num=6000):
    """
    On |z|=1 with z = exp(it):
    e^{-int} Q(e^{it}) = 2 cos(nt) - 2 cos((n-1)t) + beta
                       = -4 sin((2n-1)t/2) sin(t/2) + beta
    """
    t = np.linspace(0, 2*np.pi, num=num)
    y = 2*np.cos(n*t) - 2*np.cos((n-1)*t) + beta
    return t, y

# -----------------------------
# Scan
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for beta in (-1, 1):
        roots = q_edge_roots(n, beta)
        U = count_unimodular_roots(roots)
        rows.append({
            "n": n,
            "k": 1,
            "m": n - 1,
            "alpha": -1,
            "beta": beta,
            "degree": 2*n,
            "U_numeric": U,
            "deficit": 2*n - U,
            "all_unimodular_numeric": (U == 2*n),
            "n_mod_2": n % 2,
            "n_mod_3": n % 3,
            "n_mod_4": n % 4,
            "n_mod_6": n % 6,
            "n_mod_12": n % 12,
        })

df = pd.DataFrame(rows)

# -----------------------------
# Winner by n
# -----------------------------
winner_by_n = (
    df.sort_values(by=["n", "U_numeric", "beta"], ascending=[True, False, True])
      .groupby("n", as_index=False)
      .first()
      .reset_index(drop=True)
)

# tie detection
tie_info = (
    df.groupby("n", as_index=False)["U_numeric"]
      .max()
      .rename(columns={"U_numeric": "U_max"})
)
tie_counts = (
    df.merge(tie_info, on="n")
      .query("U_numeric == U_max")
      .groupby("n", as_index=False)
      .size()
      .rename(columns={"size": "num_betas_attaining_max"})
)
winner_by_n = winner_by_n.merge(tie_counts, on="n", how="left")

# beta comparison table
pivot = (
    df.pivot(index="n", columns="beta", values="U_numeric")
      .rename(columns={-1: "U_beta_minus", 1: "U_beta_plus"})
      .reset_index()
)
pivot["winner_beta"] = np.where(
    pivot["U_beta_minus"] > pivot["U_beta_plus"], -1,
    np.where(pivot["U_beta_minus"] < pivot["U_beta_plus"], 1, 0)
)
pivot["gap"] = np.abs(pivot["U_beta_minus"] - pivot["U_beta_plus"])
pivot["n_mod_6"] = pivot["n"] % 6
pivot["n_mod_12"] = pivot["n"] % 12

# modular summaries
mod6_summary = (
    pivot.groupby("n_mod_6", as_index=False)
         .agg(
             count=("n", "count"),
             beta_minus_wins=("winner_beta", lambda s: int(np.sum(s == -1))),
             beta_plus_wins=("winner_beta", lambda s: int(np.sum(s == 1))),
             ties=("winner_beta", lambda s: int(np.sum(s == 0))),
             avg_gap=("gap", "mean")
         )
         .sort_values("n_mod_6")
         .reset_index(drop=True)
)

mod12_summary = (
    pivot.groupby("n_mod_12", as_index=False)
         .agg(
             count=("n", "count"),
             beta_minus_wins=("winner_beta", lambda s: int(np.sum(s == -1))),
             beta_plus_wins=("winner_beta", lambda s: int(np.sum(s == 1))),
             ties=("winner_beta", lambda s: int(np.sum(s == 0))),
             avg_gap=("gap", "mean")
         )
         .sort_values("n_mod_12")
         .reset_index(drop=True)
)

# top cases
top20 = (
    df.sort_values(by=["U_numeric", "n", "beta"], ascending=[False, True, True])
      .head(20)
      .reset_index(drop=True)
)

print(f"Total cases scanned in k=1, alpha=-1 family: {len(df)}")
print(f"Distinct U values found: {sorted(df['U_numeric'].unique().tolist())}")
print(f"Min U_numeric: {int(df['U_numeric'].min())}")
print(f"Max U_numeric: {int(df['U_numeric'].max())}")
print(f"All-unimodular cases: {int(df['all_unimodular_numeric'].sum())}")

retro_table(top20, "Top 20 cases in the k=1, alpha=-1 family", max_rows=20)
retro_table(
    winner_by_n[["n","beta","U_numeric","deficit","all_unimodular_numeric","num_betas_attaining_max"]],
    "Winning beta by n",
    max_rows=80
)
retro_table(pivot.head(80), "Beta comparison table", max_rows=80)
retro_table(mod6_summary, "Winner beta by n mod 6", max_rows=10)
retro_table(mod12_summary, "Winner beta by n mod 12", max_rows=20)

# -----------------------------
# Plot 1: U for both beta values
# -----------------------------
fig1 = go.Figure()

dminus = df[df["beta"] == -1]
dplus  = df[df["beta"] == 1]

fig1.add_trace(go.Scatter(
    x=dminus["n"], y=dminus["U_numeric"],
    mode="lines+markers",
    name="beta = -1",
    line=dict(width=2.5, color=NEON_CYAN),
    marker=dict(size=6, color=NEON_CYAN)
))

fig1.add_trace(go.Scatter(
    x=dplus["n"], y=dplus["U_numeric"],
    mode="lines+markers",
    name="beta = +1",
    line=dict(width=2.5, color=NEON_AMBER),
    marker=dict(size=6, color=NEON_AMBER)
))

apply_retro(fig1, "U_numeric for Q_{n,n-1}^{-1,beta}")
fig1.update_layout(xaxis_title="n", yaxis_title="U_numeric")
fig1.show()

# -----------------------------
# Plot 2: winning beta
# -----------------------------
winner_plot = pivot.copy()
winner_plot["winner_label"] = winner_plot["winner_beta"].map({-1: "beta=-1", 1: "beta=+1", 0: "tie"})

fig2 = px.scatter(
    winner_plot,
    x="n",
    y="gap",
    color="winner_label",
    symbol="winner_label",
    title="Which beta wins, and by how much?"
)
apply_retro(fig2)
fig2.update_layout(xaxis_title="n", yaxis_title="|U(-1)-U(+1)|")
fig2.show()

# -----------------------------
# Plot 3: modular view mod 12
# FIXED: use scatter instead of strip(symbol=...)
# -----------------------------
fig3 = px.scatter(
    winner_plot,
    x="n_mod_12",
    y="gap",
    color="winner_label",
    symbol="winner_label",
    hover_data=["n", "U_beta_minus", "U_beta_plus", "winner_beta"],
    title="Winner beta grouped by n mod 12"
)
apply_retro(fig3)
fig3.update_layout(xaxis_title="n mod 12", yaxis_title="gap")
fig3.show()

# -----------------------------
# Plot 4 and 5: top case only
# -----------------------------
row = top20.iloc[0]
n0 = int(row["n"])
b0 = int(row["beta"])
U0 = int(row["U_numeric"])

roots0 = q_edge_roots(n0, b0)

theta = np.linspace(0, 2*np.pi, 800)
fig4 = go.Figure()
fig4.add_trace(go.Scatter(
    x=np.cos(theta), y=np.sin(theta),
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="unit circle"
))
fig4.add_trace(go.Scatter(
    x=np.real(roots0), y=np.imag(roots0),
    mode="markers",
    marker=dict(size=9, color=NEON_CYAN, line=dict(color=NEON_GREEN, width=1)),
    name="roots"
))
apply_retro(fig4, f"Top case roots: n={n0}, beta={b0}, U={U0}")
fig4.update_layout(
    xaxis_title="Re",
    yaxis_title="Im",
    yaxis_scaleanchor="x",
    yaxis_scaleratio=1
)
fig4.show()

t, y = trig_values(n0, b0, num=8000)
fig5 = go.Figure()
fig5.add_trace(go.Scatter(
    x=t, y=y,
    mode="lines",
    line=dict(color=NEON_AMBER, width=2),
    name="2cos(nt)-2cos((n-1)t)+beta"
))
fig5.add_trace(go.Scatter(
    x=[t.min(), t.max()], y=[0, 0],
    mode="lines",
    line=dict(color=NEON_GREEN, width=1, dash="dash"),
    name="y=0"
))
apply_retro(fig5, f"Top case trigonometric profile: n={n0}, beta={b0}")
fig5.update_layout(xaxis_title="t", yaxis_title="value")
fig5.show()

# -----------------------------
# Save CSV
# -----------------------------
if SAVE_CSV:
    csv_path = "/content/week3b_k1_alpha_minus1_scan.csv"
    df.to_csv(csv_path, index=False)
    print(f"\nSaved scan to: {csv_path}")
```

```python
# =========================================================
# WEEK 3C — Interval-by-interval zero counting
# For the focused family
#   Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#
# On the unit circle:
#   F_{n,beta}(t) = -4 sin((2n-1)t/2) sin(t/2) + beta
#
# Goal:
#   verify interval criterion for zeros of F_{n,beta}(t)
# =========================================================

import numpy as np
import pandas as pd
from math import pi
from scipy.optimize import brentq
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 120          # verification range
N_DETAIL = 60        # detailed interval table for one n
UNIT_TOL = 1e-7
EPS = 1e-12
SAVE_CSV = True

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=180, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    display(sty)

# -----------------------------
# Polynomial and trig objects
# -----------------------------
def q_edge_coeffs(n, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[1] = -1.0
    coeffs[n] = float(beta)
    coeffs[-2] = -1.0
    coeffs[-1] = 1.0
    return coeffs

def q_edge_roots(n, beta):
    return np.roots(q_edge_coeffs(n, beta))

def U_poly(n, beta, tol=UNIT_TOL):
    roots = q_edge_roots(n, beta)
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < tol))

def F(n, beta, t):
    return -4*np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0) + beta

# -----------------------------
# Interval theory objects
# -----------------------------
def interval_data(n, beta):
    """
    Intervals:
      I_j = (2πj/(2n-1), 2π(j+1)/(2n-1)), j = 0,...,2n-2
    Midpoint:
      m_j = (2j+1)π/(2n-1)
    Candidate criterion:
      2 roots in I_j iff (-1)^j = beta and 4 sin(m_j/2) > 1
    """
    rows = []
    for j in range(2*n - 1):
        left = 2*np.pi*j/(2*n - 1)
        right = 2*np.pi*(j+1)/(2*n - 1)
        mid = (2*j + 1)*np.pi/(2*n - 1)

        amp = 4*np.sin(mid/2.0)
        parity_val = 1 if (j % 2 == 0) else -1
        parity_match = (parity_val == beta)
        two_roots_pred = bool(parity_match and (amp > 1 + 1e-12))

        f_left = F(n, beta, left + EPS)
        f_mid  = F(n, beta, mid)
        f_right = F(n, beta, right - EPS)

        roots_found = []
        if two_roots_pred:
            # One root in (left, mid), one in (mid, right)
            if f_left * f_mid < 0:
                r1 = brentq(lambda t: F(n, beta, t), left + EPS, mid - EPS)
                roots_found.append(r1)
            if f_mid * f_right < 0:
                r2 = brentq(lambda t: F(n, beta, t), mid + EPS, right - EPS)
                roots_found.append(r2)

        rows.append({
            "j": j,
            "left": left,
            "mid": mid,
            "right": right,
            "parity_val": parity_val,
            "beta": beta,
            "parity_match": parity_match,
            "amp": amp,
            "amp_gt_1": bool(amp > 1 + 1e-12),
            "two_roots_pred": two_roots_pred,
            "f_left": f_left,
            "f_mid": f_mid,
            "f_right": f_right,
            "num_roots_found": len(roots_found),
            "root_1": roots_found[0] if len(roots_found) >= 1 else np.nan,
            "root_2": roots_found[1] if len(roots_found) >= 2 else np.nan,
        })

    return pd.DataFrame(rows)

def U_interval_pred(n, beta):
    d = interval_data(n, beta)
    return int(2 * d["two_roots_pred"].sum())

def U_interval_found(n, beta):
    d = interval_data(n, beta)
    return int(d["num_roots_found"].sum())

# -----------------------------
# Global verification
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for beta in (-1, 1):
        up = U_poly(n, beta)
        ui_pred = U_interval_pred(n, beta)
        ui_found = U_interval_found(n, beta)

        rows.append({
            "n": n,
            "beta": beta,
            "U_poly": up,
            "U_interval_pred": ui_pred,
            "U_interval_found": ui_found,
            "pred_matches_poly": (up == ui_pred),
            "found_matches_poly": (up == ui_found),
            "gap_pred_minus_poly": ui_pred - up,
            "gap_found_minus_poly": ui_found - up,
            "winner_label": f"beta={beta}",
            "n_mod_6": n % 6,
            "n_mod_12": n % 12,
        })

verify_df = pd.DataFrame(rows)

# winner table from interval-prediction
pivot = (
    verify_df.pivot(index="n", columns="beta", values="U_interval_pred")
    .rename(columns={-1: "U_pred_beta_minus", 1: "U_pred_beta_plus"})
    .reset_index()
)
pivot["winner_beta_pred"] = np.where(
    pivot["U_pred_beta_minus"] > pivot["U_pred_beta_plus"], -1,
    np.where(pivot["U_pred_beta_minus"] < pivot["U_pred_beta_plus"], 1, 0)
)
pivot["gap_pred"] = np.abs(pivot["U_pred_beta_minus"] - pivot["U_pred_beta_plus"])
pivot["n_mod_6"] = pivot["n"] % 6
pivot["n_mod_12"] = pivot["n"] % 12

# detailed interval table
detail_minus = interval_data(N_DETAIL, -1)
detail_plus  = interval_data(N_DETAIL, 1)

# -----------------------------
# Print summaries
# -----------------------------
print(f"Verification range: n = 2,...,{N_MAX}")
print("All pred_matches_poly:", bool(verify_df["pred_matches_poly"].all()))
print("All found_matches_poly:", bool(verify_df["found_matches_poly"].all()))

mismatch_pred = verify_df[~verify_df["pred_matches_poly"]].copy()
mismatch_found = verify_df[~verify_df["found_matches_poly"]].copy()

retro_table(
    verify_df.head(40),
    "Verification table: polynomial count vs interval prediction",
    max_rows=40
)

if len(mismatch_pred) > 0:
    retro_table(mismatch_pred, "Mismatches for predicted interval count", max_rows=40)
else:
    print("\nNo mismatches between U_poly and U_interval_pred in the tested range.")

if len(mismatch_found) > 0:
    retro_table(mismatch_found, "Mismatches for found interval roots", max_rows=40)
else:
    print("No mismatches between U_poly and U_interval_found in the tested range.")

retro_table(
    pivot.head(80),
    "Winner beta from interval criterion",
    max_rows=80
)

# -----------------------------
# Plot 1: verification of counts
# -----------------------------
fig1 = go.Figure()

for beta, color in [(-1, NEON_CYAN), (1, NEON_AMBER)]:
    dsub = verify_df[verify_df["beta"] == beta]
    fig1.add_trace(go.Scatter(
        x=dsub["n"], y=dsub["U_poly"],
        mode="lines",
        name=f"U_poly, beta={beta}",
        line=dict(color=color, width=3)
    ))
    fig1.add_trace(go.Scatter(
        x=dsub["n"], y=dsub["U_interval_pred"],
        mode="markers",
        name=f"U_interval_pred, beta={beta}",
        marker=dict(color=color, size=5, symbol="circle-open")
    ))

apply_retro(fig1, "Polynomial count vs interval prediction")
fig1.update_layout(xaxis_title="n", yaxis_title="U")
fig1.show()

# -----------------------------
# Plot 2: predicted winner beta
# -----------------------------
winner_plot = pivot.copy()
winner_plot["winner_label"] = winner_plot["winner_beta_pred"].map({-1: "beta=-1", 1: "beta=+1", 0: "tie"})

fig2 = px.scatter(
    winner_plot,
    x="n",
    y="gap_pred",
    color="winner_label",
    symbol="winner_label",
    title="Predicted winning beta from interval criterion"
)
apply_retro(fig2)
fig2.update_layout(xaxis_title="n", yaxis_title="predicted gap")
fig2.show()

# -----------------------------
# Plot 3: detailed interval amplitudes for N_DETAIL
# -----------------------------
detail_plot = detail_plus.copy()
detail_plot["beta_case"] = "beta=+1"
detail_minus_plot = detail_minus.copy()
detail_minus_plot["beta_case"] = "beta=-1"
detail_all = pd.concat([detail_minus_plot, detail_plot], ignore_index=True)

fig3 = px.scatter(
    detail_all,
    x="j",
    y="amp",
    color="beta_case",
    symbol="two_roots_pred",
    hover_data=["parity_match", "amp_gt_1", "num_roots_found"],
    title=f"Interval amplitudes for n={N_DETAIL}"
)
apply_retro(fig3)
fig3.add_hline(y=1.0, line_dash="dash", line_color=NEON_GREEN)
fig3.update_layout(xaxis_title="interval index j", yaxis_title="amp = 4 sin(mid/2)")
fig3.show()

# -----------------------------
# Tables for one detailed n
# -----------------------------
retro_table(
    detail_minus[["j","parity_val","beta","amp","parity_match","amp_gt_1","two_roots_pred","num_roots_found"]].head(80),
    f"Detailed interval table for n={N_DETAIL}, beta=-1",
    max_rows=80
)

retro_table(
    detail_plus[["j","parity_val","beta","amp","parity_match","amp_gt_1","two_roots_pred","num_roots_found"]].head(80),
    f"Detailed interval table for n={N_DETAIL}, beta=+1",
    max_rows=80
)

# -----------------------------
# Save CSVs
# -----------------------------
if SAVE_CSV:
    verify_df.to_csv("/content/week3c_interval_verification.csv", index=False)
    pivot.to_csv("/content/week3c_interval_winner_table.csv", index=False)
    detail_minus.to_csv(f"/content/week3c_detail_n{N_DETAIL}_beta_minus.csv", index=False)
    detail_plus.to_csv(f"/content/week3c_detail_n{N_DETAIL}_beta_plus.csv", index=False)
    print("\nSaved:")
    print("/content/week3c_interval_verification.csv")
    print("/content/week3c_interval_winner_table.csv")
    print(f"/content/week3c_detail_n{N_DETAIL}_beta_minus.csv")
    print(f"/content/week3c_detail_n{N_DETAIL}_beta_plus.csv")
```

```python
# =========================================================
# WEEK 3D — Critical-point criterion by interval maxima
# For the focused family
#   Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#
# On the unit circle:
#   H_n(t) = 4 sin((2n-1)t/2) sin(t/2)
# and zeros correspond to
#   H_n(t) = beta,   beta in {-1,1}
#
# Goal:
#   replace the midpoint criterion by a true interval-maximum criterion
# =========================================================

import numpy as np
import pandas as pd
from math import pi
from scipy.optimize import brentq, minimize_scalar
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 120          # global verification range
N_DETAIL = 72        # detailed interval inspection
UNIT_TOL = 1e-7
AMP_TOL = 1e-10
EPS = 1e-12
SAVE_CSV = True

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=200, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    display(sty)

# -----------------------------
# Polynomial side
# -----------------------------
def q_edge_coeffs(n, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[1] = -1.0
    coeffs[n] = float(beta)
    coeffs[-2] = -1.0
    coeffs[-1] = 1.0
    return coeffs

def q_edge_roots(n, beta):
    return np.roots(q_edge_coeffs(n, beta))

def U_poly(n, beta, tol=UNIT_TOL):
    roots = q_edge_roots(n, beta)
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < tol))

# -----------------------------
# Trigonometric side
# -----------------------------
def H(n, t):
    return 4*np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def dH(n, t):
    return 2*((2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
              + np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0))

def interval_sign(j):
    # sign of sin((2n-1)t/2) on I_j
    return 1 if (j % 2 == 0) else -1

def interval_bounds(n, j):
    left  = 2*np.pi*j/(2*n - 1)
    right = 2*np.pi*(j + 1)/(2*n - 1)
    return left, right

def midpoint(n, j):
    return (2*j + 1)*np.pi/(2*n - 1)

def midpoint_amp(n, j):
    # old heuristic amplitude
    m = midpoint(n, j)
    return 4*np.sin(m/2.0)

# -----------------------------
# Critical-point localization
# -----------------------------
def critical_point_in_interval(n, j):
    """
    Numerically maximize sign(H)*H on I_j.
    This returns the true local extremum for H on that interval.
    """
    left, right = interval_bounds(n, j)
    sgn = interval_sign(j)

    obj = lambda t: -(sgn * H(n, t))
    res = minimize_scalar(
        obj,
        bounds=(left + EPS, right - EPS),
        method="bounded",
        options={"xatol": 1e-13}
    )

    t_star = res.x
    amp_star = sgn * H(n, t_star)           # positive height
    deriv_res = dH(n, t_star)

    return t_star, amp_star, deriv_res, bool(res.success)

# -----------------------------
# Interval-by-interval table
# -----------------------------
def interval_table(n, beta):
    rows = []

    for j in range(2*n - 1):
        left, right = interval_bounds(n, j)
        mid = midpoint(n, j)
        sgn = interval_sign(j)

        amp_mid = midpoint_amp(n, j)
        t_star, amp_star, deriv_res, success = critical_point_in_interval(n, j)

        parity_match = (sgn == beta)

        # old midpoint prediction
        num_roots_mid_pred = 2 if (parity_match and amp_mid > 1 + AMP_TOL) else 0

        # new critical-amplitude prediction
        if parity_match and amp_star > 1 + AMP_TOL:
            num_roots_crit_pred = 2
        elif parity_match and abs(amp_star - 1) <= AMP_TOL:
            num_roots_crit_pred = 1
        else:
            num_roots_crit_pred = 0

        # find actual roots of H_n(t)=beta in this interval, using t_star as splitter
        eq = lambda t: H(n, t) - beta
        roots_found = []

        if num_roots_crit_pred >= 1:
            fL = eq(left + EPS)
            fM = eq(t_star)
            fR = eq(right - EPS)

            # left half
            if fL * fM < 0:
                r1 = brentq(eq, left + EPS, t_star)
                roots_found.append(r1)
            elif abs(fM) <= 1e-10:
                roots_found.append(t_star)

            # right half
            if fM * fR < 0:
                r2 = brentq(eq, t_star, right - EPS)
                # avoid double insertion if tangent at t_star
                if len(roots_found) == 0 or abs(r2 - roots_found[-1]) > 1e-8:
                    roots_found.append(r2)

        rows.append({
            "j": j,
            "left": left,
            "mid": mid,
            "right": right,
            "sign_H_on_interval": sgn,
            "beta": beta,
            "parity_match": parity_match,
            "amp_mid": amp_mid,
            "amp_star": amp_star,
            "amp_gain": amp_star - amp_mid,
            "t_star": t_star,
            "deriv_residual_at_t_star": deriv_res,
            "optimizer_success": success,
            "num_roots_mid_pred": num_roots_mid_pred,
            "num_roots_crit_pred": num_roots_crit_pred,
            "num_roots_found": len(roots_found),
            "root_1": roots_found[0] if len(roots_found) >= 1 else np.nan,
            "root_2": roots_found[1] if len(roots_found) >= 2 else np.nan,
        })

    return pd.DataFrame(rows)

def U_mid_pred(n, beta):
    d = interval_table(n, beta)
    return int(d["num_roots_mid_pred"].sum())

def U_crit_pred(n, beta):
    d = interval_table(n, beta)
    return int(d["num_roots_crit_pred"].sum())

def U_crit_found(n, beta):
    d = interval_table(n, beta)
    return int(d["num_roots_found"].sum())

# -----------------------------
# Global verification
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    for beta in (-1, 1):
        up = U_poly(n, beta)
        um = U_mid_pred(n, beta)
        uc = U_crit_pred(n, beta)
        uf = U_crit_found(n, beta)

        rows.append({
            "n": n,
            "beta": beta,
            "U_poly": up,
            "U_mid_pred": um,
            "U_crit_pred": uc,
            "U_crit_found": uf,
            "mid_matches_poly": (up == um),
            "crit_pred_matches_poly": (up == uc),
            "crit_found_matches_poly": (up == uf),
            "gap_mid_minus_poly": um - up,
            "gap_crit_pred_minus_poly": uc - up,
            "gap_crit_found_minus_poly": uf - up,
        })

verify_df = pd.DataFrame(rows)

# winner table from critical criterion
pivot = (
    verify_df.pivot(index="n", columns="beta", values="U_crit_pred")
    .rename(columns={-1: "U_crit_beta_minus", 1: "U_crit_beta_plus"})
    .reset_index()
)
pivot["winner_beta_crit"] = np.where(
    pivot["U_crit_beta_minus"] > pivot["U_crit_beta_plus"], -1,
    np.where(pivot["U_crit_beta_minus"] < pivot["U_crit_beta_plus"], 1, 0)
)
pivot["gap_crit"] = np.abs(pivot["U_crit_beta_minus"] - pivot["U_crit_beta_plus"])
pivot["n_mod_6"] = pivot["n"] % 6
pivot["n_mod_12"] = pivot["n"] % 12

# detailed tables for one chosen n
detail_minus = interval_table(N_DETAIL, -1)
detail_plus  = interval_table(N_DETAIL, 1)

# -----------------------------
# Summary output
# -----------------------------
print(f"Verification range: n = 2,...,{N_MAX}")
print("All mid_matches_poly:", bool(verify_df["mid_matches_poly"].all()))
print("All crit_pred_matches_poly:", bool(verify_df["crit_pred_matches_poly"].all()))
print("All crit_found_matches_poly:", bool(verify_df["crit_found_matches_poly"].all()))

mismatch_mid = verify_df[~verify_df["mid_matches_poly"]].copy()
mismatch_crit_pred = verify_df[~verify_df["crit_pred_matches_poly"]].copy()
mismatch_crit_found = verify_df[~verify_df["crit_found_matches_poly"]].copy()

retro_table(
    verify_df.head(40),
    "Verification table: polynomial count vs midpoint vs critical criterion",
    max_rows=40
)

if len(mismatch_mid) > 0:
    retro_table(mismatch_mid, "Midpoint-criterion mismatches", max_rows=40)
else:
    print("\nNo midpoint mismatches in the tested range.")

if len(mismatch_crit_pred) > 0:
    retro_table(mismatch_crit_pred, "Critical-criterion mismatches", max_rows=40)
else:
    print("No critical-prediction mismatches in the tested range.")

if len(mismatch_crit_found) > 0:
    retro_table(mismatch_crit_found, "Root-finding mismatches under critical criterion", max_rows=40)
else:
    print("No root-finding mismatches under critical criterion in the tested range.")

retro_table(
    pivot.head(80),
    "Winner beta from critical-interval criterion",
    max_rows=80
)

# -----------------------------
# Plot 1: compare three counts
# -----------------------------
fig1 = go.Figure()

for beta, color in [(-1, NEON_CYAN), (1, NEON_AMBER)]:
    dsub = verify_df[verify_df["beta"] == beta]
    fig1.add_trace(go.Scatter(
        x=dsub["n"], y=dsub["U_poly"],
        mode="lines",
        name=f"U_poly, beta={beta}",
        line=dict(color=color, width=3)
    ))
    fig1.add_trace(go.Scatter(
        x=dsub["n"], y=dsub["U_mid_pred"],
        mode="markers",
        name=f"U_mid_pred, beta={beta}",
        marker=dict(color=color, size=5, symbol="x")
    ))
    fig1.add_trace(go.Scatter(
        x=dsub["n"], y=dsub["U_crit_pred"],
        mode="markers",
        name=f"U_crit_pred, beta={beta}",
        marker=dict(color=color, size=6, symbol="circle-open")
    ))

apply_retro(fig1, "Polynomial count vs midpoint vs critical criterion")
fig1.update_layout(xaxis_title="n", yaxis_title="U")
fig1.show()

# -----------------------------
# Plot 2: critical winner beta
# -----------------------------
winner_plot = pivot.copy()
winner_plot["winner_label"] = winner_plot["winner_beta_crit"].map({-1: "beta=-1", 1: "beta=+1", 0: "tie"})

fig2 = px.scatter(
    winner_plot,
    x="n",
    y="gap_crit",
    color="winner_label",
    symbol="winner_label",
    title="Winner beta from critical-interval criterion"
)
apply_retro(fig2)
fig2.update_layout(xaxis_title="n", yaxis_title="critical gap")
fig2.show()

# -----------------------------
# Plot 3: midpoint amp vs critical amp for N_DETAIL
# -----------------------------
detail_minus_plot = detail_minus.copy()
detail_minus_plot["beta_case"] = "beta=-1"
detail_plus_plot = detail_plus.copy()
detail_plus_plot["beta_case"] = "beta=+1"
detail_all = pd.concat([detail_minus_plot, detail_plus_plot], ignore_index=True)

fig3 = go.Figure()
for beta_case, color in [("beta=-1", NEON_CYAN), ("beta=+1", NEON_AMBER)]:
    dsub = detail_all[detail_all["beta_case"] == beta_case]
    fig3.add_trace(go.Scatter(
        x=dsub["j"], y=dsub["amp_mid"],
        mode="lines+markers",
        name=f"amp_mid, {beta_case}",
        line=dict(color=color, width=1.5, dash="dot"),
        marker=dict(size=4, color=color)
    ))
    fig3.add_trace(go.Scatter(
        x=dsub["j"], y=dsub["amp_star"],
        mode="lines+markers",
        name=f"amp_star, {beta_case}",
        line=dict(color=color, width=2.5),
        marker=dict(size=4, color=color)
    ))

apply_retro(fig3, f"Midpoint amplitude vs critical amplitude for n={N_DETAIL}")
fig3.add_hline(y=1.0, line_dash="dash", line_color=NEON_GREEN)
fig3.update_layout(xaxis_title="interval index j", yaxis_title="amplitude")
fig3.show()

# -----------------------------
# Tables for chosen n
# -----------------------------
retro_table(
    detail_minus[[
        "j","sign_H_on_interval","beta","amp_mid","amp_star","amp_gain",
        "num_roots_mid_pred","num_roots_crit_pred","num_roots_found",
        "deriv_residual_at_t_star"
    ]].head(100),
    f"Detailed interval table for n={N_DETAIL}, beta=-1",
    max_rows=100
)

retro_table(
    detail_plus[[
        "j","sign_H_on_interval","beta","amp_mid","amp_star","amp_gain",
        "num_roots_mid_pred","num_roots_crit_pred","num_roots_found",
        "deriv_residual_at_t_star"
    ]].head(100),
    f"Detailed interval table for n={N_DETAIL}, beta=+1",
    max_rows=100
)

# -----------------------------
# Save CSVs
# -----------------------------
if SAVE_CSV:
    verify_df.to_csv("/content/week3d_critical_verification.csv", index=False)
    pivot.to_csv("/content/week3d_critical_winner_table.csv", index=False)
    detail_minus.to_csv(f"/content/week3d_detail_n{N_DETAIL}_beta_minus.csv", index=False)
    detail_plus.to_csv(f"/content/week3d_detail_n{N_DETAIL}_beta_plus.csv", index=False)
    print("\nSaved:")
    print("/content/week3d_critical_verification.csv")
    print("/content/week3d_critical_winner_table.csv")
    print(f"/content/week3d_detail_n{N_DETAIL}_beta_minus.csv")
    print(f"/content/week3d_detail_n{N_DETAIL}_beta_plus.csv")
```

```python
# =========================================================
# WEEK 3E — Uniqueness of the interval critical point
# For
#   H_n(t) = 4 sin((2n-1)t/2) sin(t/2)
#
# Goal:
#   numerically verify that on each interval
#   I_{n,j} = (2πj/(2n-1), 2π(j+1)/(2n-1))
#   the signed function (-1)^j H_n(t)
#   has a unique interior critical point,
#   and that this point is the interval maximum.
# =========================================================

import numpy as np
import pandas as pd
from scipy.optimize import brentq, minimize_scalar
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 120          # verification range
N_DETAIL = 72        # detailed inspection
SUBDIV = 160         # derivative-root detection grid per interval
EPS = 1e-12
SAVE_CSV = True

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=200, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    display(sty)

# -----------------------------
# H_n and derivatives
# -----------------------------
def H(n, t):
    return 4*np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def dH(n, t):
    return 2*((2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
              + np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0))

def interval_bounds(n, j):
    left = 2*np.pi*j/(2*n - 1)
    right = 2*np.pi*(j + 1)/(2*n - 1)
    return left, right

def interval_sign(j):
    return 1 if (j % 2 == 0) else -1

# -----------------------------
# Root finding for dH on one interval
# -----------------------------
def derivative_roots_in_interval(n, j, subdiv=SUBDIV, eps=EPS):
    a, b = interval_bounds(n, j)
    xs = np.linspace(a + eps, b - eps, subdiv + 1)
    ys = dH(n, xs)

    roots = []
    for k in range(subdiv):
        x1, x2 = xs[k], xs[k + 1]
        y1, y2 = ys[k], ys[k + 1]

        if (not np.isfinite(y1)) or (not np.isfinite(y2)):
            continue

        if abs(y1) < 1e-12:
            r = x1
        elif y1 * y2 < 0:
            try:
                r = brentq(lambda t: dH(n, t), x1, x2)
            except ValueError:
                continue
        else:
            continue

        if len(roots) == 0 or abs(r - roots[-1]) > 1e-8:
            roots.append(r)

    return roots

# -----------------------------
# Interval critical-point data
# -----------------------------
def interval_critical_data(n, j):
    a, b = interval_bounds(n, j)
    mid = 0.5*(a + b)
    s = interval_sign(j)

    G = lambda t: s * H(n, t)
    dG = lambda t: s * dH(n, t)

    # optimizer-based maximum
    res = minimize_scalar(
        lambda t: -G(t),
        bounds=(a + EPS, b - EPS),
        method="bounded",
        options={"xatol": 1e-13}
    )
    t_star = res.x
    G_star = G(t_star)

    # derivative roots
    crit_roots = derivative_roots_in_interval(n, j, subdiv=SUBDIV, eps=EPS)
    num_crit_roots = len(crit_roots)

    # best derivative root if it exists
    if num_crit_roots > 0:
        t_crit = crit_roots[np.argmax([G(r) for r in crit_roots])]
        G_crit = G(t_crit)
        delta_optimizer_vs_root = abs(t_star - t_crit)
    else:
        t_crit = np.nan
        G_crit = np.nan
        delta_optimizer_vs_root = np.nan

    # derivative sign near endpoints
    dG_left = dG(a + 1e-8*(b-a))
    dG_right = dG(b - 1e-8*(b-a))

    # rough monotonicity pattern expected:
    # increasing from left, decreasing to right
    left_positive = (dG_left > 0)
    right_negative = (dG_right < 0)

    # normalized critical location in interval
    rel_pos = (t_star - a)/(b - a)

    return {
        "n": n,
        "j": j,
        "a": a,
        "mid": mid,
        "b": b,
        "sign_interval": s,
        "num_crit_roots": num_crit_roots,
        "t_star_optimizer": t_star,
        "t_star_derivative": t_crit,
        "G_star_optimizer": G_star,
        "G_star_derivative": G_crit,
        "delta_optimizer_vs_root": delta_optimizer_vs_root,
        "left_positive": left_positive,
        "right_negative": right_negative,
        "dG_left": dG_left,
        "dG_right": dG_right,
        "optimizer_success": bool(res.success),
        "rel_pos_in_interval": rel_pos,
        "delta_from_mid": t_star - mid,
    }

def full_interval_table(n):
    return pd.DataFrame([interval_critical_data(n, j) for j in range(2*n - 1)])

# -----------------------------
# Global verification
# -----------------------------
summary_rows = []
all_failures = []

for n in range(2, N_MAX + 1):
    table_n = full_interval_table(n)

    num_bad_count = int((table_n["num_crit_roots"] != 1).sum())
    num_bad_left = int((~table_n["left_positive"]).sum())
    num_bad_right = int((~table_n["right_negative"]).sum())
    num_bad_match = int((table_n["delta_optimizer_vs_root"] > 1e-6).sum())

    summary_rows.append({
        "n": n,
        "num_intervals": len(table_n),
        "bad_num_crit_roots": num_bad_count,
        "bad_left_sign": num_bad_left,
        "bad_right_sign": num_bad_right,
        "bad_optimizer_vs_root": num_bad_match,
        "all_good_for_n": (num_bad_count == 0 and num_bad_left == 0 and num_bad_right == 0 and num_bad_match == 0),
        "max_delta_optimizer_vs_root": np.nanmax(table_n["delta_optimizer_vs_root"].values),
        "min_rel_pos": np.nanmin(table_n["rel_pos_in_interval"].values),
        "max_rel_pos": np.nanmax(table_n["rel_pos_in_interval"].values),
    })

    if not (num_bad_count == 0 and num_bad_left == 0 and num_bad_right == 0 and num_bad_match == 0):
        bad = table_n[
            (table_n["num_crit_roots"] != 1) |
            (~table_n["left_positive"]) |
            (~table_n["right_negative"]) |
            (table_n["delta_optimizer_vs_root"] > 1e-6)
        ].copy()
        all_failures.append(bad)

summary_df = pd.DataFrame(summary_rows)

all_intervals_have_unique_critical = bool((summary_df["bad_num_crit_roots"] == 0).all())
all_left_signs_ok = bool((summary_df["bad_left_sign"] == 0).all())
all_right_signs_ok = bool((summary_df["bad_right_sign"] == 0).all())
all_optimizer_matches_root = bool((summary_df["bad_optimizer_vs_root"] == 0).all())
all_good = bool(summary_df["all_good_for_n"].all())

detail_df = full_interval_table(N_DETAIL)

print(f"Verification range: n = 2,...,{N_MAX}")
print("All intervals have exactly one critical root:", all_intervals_have_unique_critical)
print("All signed derivatives are positive near left endpoints:", all_left_signs_ok)
print("All signed derivatives are negative near right endpoints:", all_right_signs_ok)
print("Optimizer critical point matches derivative root:", all_optimizer_matches_root)
print("All interval tests passed:", all_good)

retro_table(summary_df.head(40), "Summary by n: interval critical-point verification", max_rows=40)

if all_failures:
    failures_df = pd.concat(all_failures, ignore_index=True)
    retro_table(failures_df.head(80), "Failure table (if any)", max_rows=80)
else:
    print("\nNo failures detected in the tested range.")

retro_table(
    detail_df[[
        "j","sign_interval","num_crit_roots","t_star_optimizer","t_star_derivative",
        "G_star_optimizer","delta_optimizer_vs_root",
        "left_positive","right_negative","rel_pos_in_interval","delta_from_mid"
    ]].head(120),
    f"Detailed critical-point table for n={N_DETAIL}",
    max_rows=120
)

# -----------------------------
# Plot 1: summary by n
# -----------------------------
fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=summary_df["n"],
    y=summary_df["bad_num_crit_roots"],
    mode="lines+markers",
    name="bad_num_crit_roots",
    line=dict(color=NEON_CYAN, width=2),
    marker=dict(size=5, color=NEON_CYAN)
))

fig1.add_trace(go.Scatter(
    x=summary_df["n"],
    y=summary_df["bad_left_sign"] + summary_df["bad_right_sign"],
    mode="lines+markers",
    name="bad_endpoint_signs",
    line=dict(color=NEON_AMBER, width=2),
    marker=dict(size=5, color=NEON_AMBER)
))

fig1.add_trace(go.Scatter(
    x=summary_df["n"],
    y=summary_df["bad_optimizer_vs_root"],
    mode="lines+markers",
    name="bad_optimizer_vs_root",
    line=dict(color=NEON_PINK, width=2),
    marker=dict(size=5, color=NEON_PINK)
))

apply_retro(fig1, "Week 3E summary: critical-point uniqueness checks")
fig1.update_layout(xaxis_title="n", yaxis_title="number of bad intervals")
fig1.show()

# -----------------------------
# Plot 2: critical location inside interval
# -----------------------------
fig2 = px.scatter(
    detail_df,
    x="j",
    y="rel_pos_in_interval",
    color="sign_interval",
    title=f"Relative critical-point location inside interval for n={N_DETAIL}"
)
apply_retro(fig2)
fig2.update_layout(xaxis_title="interval index j", yaxis_title="(t* - a)/(b - a)")
fig2.show()

# -----------------------------
# Plot 3: delta from midpoint
# -----------------------------
fig3 = px.scatter(
    detail_df,
    x="j",
    y="delta_from_mid",
    color="sign_interval",
    title=f"Critical point minus midpoint for n={N_DETAIL}"
)
apply_retro(fig3)
fig3.update_layout(xaxis_title="interval index j", yaxis_title="t* - midpoint")
fig3.show()

# -----------------------------
# Plot 4: maximum value by interval
# -----------------------------
fig4 = px.scatter(
    detail_df,
    x="j",
    y="G_star_optimizer",
    color="sign_interval",
    title=f"Signed interval maximum for n={N_DETAIL}"
)
apply_retro(fig4)
fig4.update_layout(xaxis_title="interval index j", yaxis_title="max of signed H_n on interval")
fig4.show()

# -----------------------------
# Save CSVs
# -----------------------------
if SAVE_CSV:
    summary_df.to_csv("/content/week3e_critical_uniqueness_summary.csv", index=False)
    detail_df.to_csv(f"/content/week3e_critical_uniqueness_detail_n{N_DETAIL}.csv", index=False)
    print("\nSaved:")
    print("/content/week3e_critical_uniqueness_summary.csv")
    print(f"/content/week3e_critical_uniqueness_detail_n{N_DETAIL}.csv")
```

```python
# =========================================================
# WEEK 3F — Final computational closure for Week 3
#
# Family:
#   Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#
# Trig object:
#   H_n(t) = 4 sin((2n-1)t/2) sin(t/2)
#
# Goals:
#   1) verify critical-interval counting up to n <= N_MAX
#   2) compare midpoint vs critical criterion
#   3) test beta_opt(n) = (-1)^floor((n+1)/6)
#   4) export paper-ready tables and figures
# =========================================================

import os
import numpy as np
import pandas as pd
from scipy.optimize import brentq
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 200
N_DETAIL = 72
UNIT_TOL = 1e-7
EPS = 1e-10
LEVEL_TOL = 1e-10
OUTDIR = "/content/week3f_outputs"
SAVE_FILES = True

os.makedirs(OUTDIR, exist_ok=True)

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=220, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    display(sty)

def save_plotly_html(fig, filename):
    path = os.path.join(OUTDIR, filename)
    fig.write_html(path, include_plotlyjs="cdn")
    return path

# -----------------------------
# Polynomial side
# -----------------------------
def q_edge_coeffs(n, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[1] = -1.0
    coeffs[n] = float(beta)
    coeffs[-2] = -1.0
    coeffs[-1] = 1.0
    return coeffs

def U_poly(n, beta, tol=UNIT_TOL):
    roots = np.roots(q_edge_coeffs(n, beta))
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < tol))

# -----------------------------
# Trigonometric side
# -----------------------------
def H(n, t):
    return 4*np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def dH(n, t):
    return 2*((2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
              + np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0))

def interval_bounds(n, j):
    a = 2*np.pi*j/(2*n - 1)
    b = 2*np.pi*(j + 1)/(2*n - 1)
    return a, b

def interval_sign(j):
    return 1 if (j % 2 == 0) else -1

def midpoint(n, j):
    a, b = interval_bounds(n, j)
    return 0.5*(a + b)

def midpoint_amp(n, j):
    return 4*np.sin(midpoint(n, j)/2.0)

def beta_formula(n):
    # beta_opt(n) = (-1)^floor((n+1)/6)
    return 1 if (((n + 1) // 6) % 2 == 0) else -1

# -----------------------------
# Interval data for one n
# -----------------------------
def interval_max_table(n):
    rows = []
    for j in range(2*n - 1):
        a, b = interval_bounds(n, j)
        s = interval_sign(j)

        # signed derivative should go + to - on the interval
        dG = lambda t: s * dH(n, t)
        left_test = dG(a + EPS*(b-a))
        right_test = dG(b - EPS*(b-a))

        # critical root by brentq
        t_star = brentq(dG, a + EPS*(b-a), b - EPS*(b-a))
        amp_star = s * H(n, t_star)
        amp_mid = midpoint_amp(n, j)

        rows.append({
            "n": n,
            "j": j,
            "a": a,
            "b": b,
            "mid": 0.5*(a+b),
            "sign_interval": s,
            "amp_mid": amp_mid,
            "amp_star": amp_star,
            "amp_gain": amp_star - amp_mid,
            "t_star": t_star,
            "rel_pos_in_interval": (t_star - a)/(b - a),
            "delta_from_mid": t_star - 0.5*(a+b),
            "left_dG_positive": bool(left_test > 0),
            "right_dG_negative": bool(right_test < 0),
        })
    return pd.DataFrame(rows)

# -----------------------------
# Count from interval maxima
# -----------------------------
def interval_count_from_amplitudes(interval_df, beta, amp_col):
    d = interval_df.copy()
    # sign must match beta; then compare amplitude to 1
    count = 0
    for _, row in d.iterrows():
        if row["sign_interval"] != beta:
            continue
        amp = row[amp_col]
        if amp > 1 + LEVEL_TOL:
            count += 2
        elif abs(amp - 1) <= LEVEL_TOL:
            count += 1
    return int(count)

# -----------------------------
# Global scan
# -----------------------------
summary_rows = []
detail_tables = {}

for n in range(2, N_MAX + 1):
    interval_df = interval_max_table(n)

    if n == N_DETAIL:
        detail_tables["detail"] = interval_df.copy()

    U_mid_minus = interval_count_from_amplitudes(interval_df, -1, "amp_mid")
    U_mid_plus  = interval_count_from_amplitudes(interval_df,  1, "amp_mid")
    U_crit_minus = interval_count_from_amplitudes(interval_df, -1, "amp_star")
    U_crit_plus  = interval_count_from_amplitudes(interval_df,  1, "amp_star")

    U_poly_minus = U_poly(n, -1)
    U_poly_plus  = U_poly(n,  1)

    winner_beta_crit = -1 if U_crit_minus > U_crit_plus else (1 if U_crit_plus > U_crit_minus else 0)
    winner_beta_poly = -1 if U_poly_minus > U_poly_plus else (1 if U_poly_plus > U_poly_minus else 0)
    winner_beta_formula = beta_formula(n)

    summary_rows.append({
        "n": n,
        "U_poly_minus": U_poly_minus,
        "U_poly_plus": U_poly_plus,
        "U_mid_minus": U_mid_minus,
        "U_mid_plus": U_mid_plus,
        "U_crit_minus": U_crit_minus,
        "U_crit_plus": U_crit_plus,
        "mid_minus_matches": (U_mid_minus == U_poly_minus),
        "mid_plus_matches": (U_mid_plus == U_poly_plus),
        "crit_minus_matches": (U_crit_minus == U_poly_minus),
        "crit_plus_matches": (U_crit_plus == U_poly_plus),
        "winner_beta_poly": winner_beta_poly,
        "winner_beta_crit": winner_beta_crit,
        "winner_beta_formula": winner_beta_formula,
        "formula_matches_crit": (winner_beta_formula == winner_beta_crit),
        "formula_matches_poly": (winner_beta_formula == winner_beta_poly),
        "gap_poly": abs(U_poly_minus - U_poly_plus),
        "gap_crit": abs(U_crit_minus - U_crit_plus),
        "n_mod_6": n % 6,
        "n_mod_12": n % 12,
    })

summary_df = pd.DataFrame(summary_rows)

# -----------------------------
# Paper-ready derived tables
# -----------------------------
midpoint_failures = summary_df[
    (~summary_df["mid_minus_matches"]) | (~summary_df["mid_plus_matches"])
].copy()

critical_failures = summary_df[
    (~summary_df["crit_minus_matches"]) | (~summary_df["crit_plus_matches"])
].copy()

formula_failures = summary_df[
    (~summary_df["formula_matches_crit"])
].copy()

winner_by_n = summary_df[[
    "n", "U_crit_minus", "U_crit_plus", "winner_beta_crit", "gap_crit",
    "winner_beta_formula", "formula_matches_crit", "n_mod_6", "n_mod_12"
]].copy()

mod12_summary = (
    summary_df.groupby("n_mod_12", as_index=False)
    .agg(
        count=("n", "count"),
        beta_minus_wins=("winner_beta_crit", lambda s: int(np.sum(s == -1))),
        beta_plus_wins=("winner_beta_crit", lambda s: int(np.sum(s == 1))),
        ties=("winner_beta_crit", lambda s: int(np.sum(s == 0))),
        avg_gap=("gap_crit", "mean")
    )
    .sort_values("n_mod_12")
    .reset_index(drop=True)
)

verification_summary = pd.DataFrame([{
    "N_MAX": N_MAX,
    "all_crit_minus_matches": bool(summary_df["crit_minus_matches"].all()),
    "all_crit_plus_matches": bool(summary_df["crit_plus_matches"].all()),
    "all_critical_matches": bool(summary_df["crit_minus_matches"].all() and summary_df["crit_plus_matches"].all()),
    "all_formula_matches_crit": bool(summary_df["formula_matches_crit"].all()),
    "all_critical_gaps_equal_2": bool((summary_df["gap_crit"] == 2).all()),
    "num_midpoint_failures": int(len(midpoint_failures)),
    "num_critical_failures": int(len(critical_failures)),
    "num_formula_failures": int(len(formula_failures)),
}])

# -----------------------------
# Console summary
# -----------------------------
print(f"Verification range: n = 2,...,{N_MAX}")
print("All crit_minus_matches:", bool(summary_df["crit_minus_matches"].all()))
print("All crit_plus_matches:", bool(summary_df["crit_plus_matches"].all()))
print("All critical matches:", bool(summary_df["crit_minus_matches"].all() and summary_df["crit_plus_matches"].all()))
print("All beta_opt_formula matches critical winner:", bool(summary_df["formula_matches_crit"].all()))
print("All critical gaps equal 2:", bool((summary_df["gap_crit"] == 2).all()))
print("Number of midpoint failures:", len(midpoint_failures))
print("Number of critical failures:", len(critical_failures))
print("Number of formula mismatches:", len(formula_failures))

# -----------------------------
# Show core paper tables
# -----------------------------
retro_table(verification_summary, "Verification summary for Week 3F", max_rows=5)
retro_table(midpoint_failures, "Midpoint failures", max_rows=40)
retro_table(critical_failures if len(critical_failures) else pd.DataFrame({"message": ["No critical failures detected."]}),
            "Critical-criterion failures", max_rows=10)
retro_table(formula_failures if len(formula_failures) else pd.DataFrame({"message": ["No formula mismatches detected."]}),
            "Formula mismatches", max_rows=10)
retro_table(winner_by_n.head(80), "Winning beta by n", max_rows=80)
retro_table(mod12_summary, "Winner beta by n mod 12", max_rows=20)

# -----------------------------
# Figures
# -----------------------------
# Figure 1: U_poly vs U_crit
fig1 = go.Figure()
fig1.add_trace(go.Scatter(
    x=summary_df["n"], y=summary_df["U_poly_minus"],
    mode="lines", name="U_poly, beta=-1",
    line=dict(color=NEON_CYAN, width=3)
))
fig1.add_trace(go.Scatter(
    x=summary_df["n"], y=summary_df["U_crit_minus"],
    mode="markers", name="U_crit, beta=-1",
    marker=dict(color=NEON_CYAN, size=5, symbol="circle-open")
))
fig1.add_trace(go.Scatter(
    x=summary_df["n"], y=summary_df["U_poly_plus"],
    mode="lines", name="U_poly, beta=+1",
    line=dict(color=NEON_AMBER, width=3)
))
fig1.add_trace(go.Scatter(
    x=summary_df["n"], y=summary_df["U_crit_plus"],
    mode="markers", name="U_crit, beta=+1",
    marker=dict(color=NEON_AMBER, size=5, symbol="circle-open")
))
apply_retro(fig1, "Polynomial count vs critical-interval count")
fig1.update_layout(xaxis_title="n", yaxis_title="U")
fig1.show()

# Figure 2: winner beta vs n
winner_plot = winner_by_n.copy()
winner_plot["winner_label"] = winner_plot["winner_beta_crit"].map({-1: "beta=-1", 1: "beta=+1", 0: "tie"})
fig2 = px.scatter(
    winner_plot,
    x="n",
    y="gap_crit",
    color="winner_label",
    symbol="winner_label",
    title="Winning beta and critical gap"
)
apply_retro(fig2)
fig2.update_layout(xaxis_title="n", yaxis_title="gap")
fig2.show()

# Figure 3: mod 12 summary
mod12_plot = mod12_summary.melt(
    id_vars=["n_mod_12", "count", "avg_gap"],
    value_vars=["beta_minus_wins", "beta_plus_wins", "ties"],
    var_name="category",
    value_name="value"
)
fig3 = px.bar(
    mod12_plot,
    x="n_mod_12",
    y="value",
    color="category",
    barmode="group",
    title="Winner counts by residue class mod 12"
)
apply_retro(fig3)
fig3.update_layout(xaxis_title="n mod 12", yaxis_title="count")
fig3.show()

# Figure 4: detail midpoint vs critical amplitudes
detail_df = detail_tables["detail"].copy()
fig4 = go.Figure()
fig4.add_trace(go.Scatter(
    x=detail_df["j"], y=detail_df["amp_mid"],
    mode="lines+markers",
    name="amp_mid",
    line=dict(color=NEON_CYAN, width=2, dash="dot"),
    marker=dict(size=4, color=NEON_CYAN)
))
fig4.add_trace(go.Scatter(
    x=detail_df["j"], y=detail_df["amp_star"],
    mode="lines+markers",
    name="amp_star",
    line=dict(color=NEON_AMBER, width=3),
    marker=dict(size=4, color=NEON_AMBER)
))
fig4.add_hline(y=1.0, line_dash="dash", line_color=NEON_GREEN)
apply_retro(fig4, f"Midpoint vs critical amplitudes for n={N_DETAIL}")
fig4.update_layout(xaxis_title="interval index j", yaxis_title="amplitude")
fig4.show()

# Figure 5: critical point location inside interval
fig5 = px.scatter(
    detail_df,
    x="j",
    y="rel_pos_in_interval",
    color="sign_interval",
    title=f"Relative critical-point location for n={N_DETAIL}"
)
apply_retro(fig5)
fig5.update_layout(xaxis_title="interval index j", yaxis_title="(t* - a)/(b - a)")
fig5.show()

# -----------------------------
# Save outputs
# -----------------------------
if SAVE_FILES:
    summary_df.to_csv(os.path.join(OUTDIR, "week3f_full_summary.csv"), index=False)
    verification_summary.to_csv(os.path.join(OUTDIR, "week3f_verification_summary.csv"), index=False)
    midpoint_failures.to_csv(os.path.join(OUTDIR, "week3f_midpoint_failures.csv"), index=False)
    critical_failures.to_csv(os.path.join(OUTDIR, "week3f_critical_failures.csv"), index=False)
    formula_failures.to_csv(os.path.join(OUTDIR, "week3f_formula_failures.csv"), index=False)
    winner_by_n.to_csv(os.path.join(OUTDIR, "week3f_winner_by_n.csv"), index=False)
    mod12_summary.to_csv(os.path.join(OUTDIR, "week3f_mod12_summary.csv"), index=False)
    detail_df.to_csv(os.path.join(OUTDIR, f"week3f_detail_n{N_DETAIL}.csv"), index=False)

    save_plotly_html(fig1, "figure1_poly_vs_critical.html")
    save_plotly_html(fig2, "figure2_winner_beta_gap.html")
    save_plotly_html(fig3, "figure3_mod12_summary.html")
    save_plotly_html(fig4, "figure4_mid_vs_critical_amplitudes.html")
    save_plotly_html(fig5, "figure5_critical_location.html")

    print("\nSaved outputs in:")
    print(OUTDIR)
    print("\nCSV files:")
    print("- week3f_full_summary.csv")
    print("- week3f_verification_summary.csv")
    print("- week3f_midpoint_failures.csv")
    print("- week3f_critical_failures.csv")
    print("- week3f_formula_failures.csv")
    print("- week3f_winner_by_n.csv")
    print("- week3f_mod12_summary.csv")
    print(f"- week3f_detail_n{N_DETAIL}.csv")
    print("\nHTML figures:")
    print("- figure1_poly_vs_critical.html")
    print("- figure2_winner_beta_gap.html")
    print("- figure3_mod12_summary.html")
    print("- figure4_mid_vs_critical_amplitudes.html")
    print("- figure5_critical_location.html")
```

```python
# =========================================================
# WEEK 3G — Empirical structure of the winning beta
# Focus only on:
#   Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#
# Goal:
#   understand the real winner-beta pattern before Week 4
# =========================================================

import os
import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MAX = 240
UNIT_TOL = 1e-7
OUTDIR = "/content/week3g_outputs"
SAVE_FILES = True
os.makedirs(OUTDIR, exist_ok=True)

# -----------------------------
# Retro palette
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"
NEON_ORANGE = "#FF6A3D"

# -----------------------------
# Styling helpers
# -----------------------------
def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            x=1.02,
            y=1.0,
            xanchor="left",
            yanchor="top",
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=60, r=220, t=80, b=60),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {
                "selector": "table",
                "props": [
                    ("background-color", BG),
                    ("color", TEXT),
                    ("border-collapse", "collapse"),
                    ("font-family", "Courier New, monospace")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#111111"),
                    ("color", NEON_AMBER),
                    ("border", "1px solid #333333"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "td",
                "props": [
                    ("background-color", "#050505"),
                    ("color", TEXT),
                    ("border", "1px solid #222222"),
                    ("padding", "6px")
                ]
            },
            {
                "selector": "caption",
                "props": [
                    ("color", NEON_GREEN),
                    ("font-size", "16px"),
                    ("font-weight", "bold"),
                    ("caption-side", "top"),
                    ("padding", "8px")
                ]
            },
        ])
    )
    display(sty)

# -----------------------------
# Family
# -----------------------------
def q_edge_coeffs(n, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[1] = -1.0
    coeffs[n] = float(beta)
    coeffs[-2] = -1.0
    coeffs[-1] = 1.0
    return coeffs

def U_poly(n, beta, tol=UNIT_TOL):
    roots = np.roots(q_edge_coeffs(n, beta))
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < tol))

# -----------------------------
# Candidate block laws
# -----------------------------
def beta_formula_shift(n, shift):
    # family of simple 6-block laws:
    # beta = (-1)^floor((n+shift)/6)
    return 1 if (((n + shift) // 6) % 2 == 0) else -1

# -----------------------------
# Main scan
# -----------------------------
rows = []
for n in range(2, N_MAX + 1):
    U_minus = U_poly(n, -1)
    U_plus  = U_poly(n, 1)
    winner = -1 if U_minus > U_plus else (1 if U_plus > U_minus else 0)
    gap = abs(U_minus - U_plus)

    row = {
        "n": n,
        "U_minus": U_minus,
        "U_plus": U_plus,
        "winner_beta": winner,
        "gap": gap,
        "n_mod_6": n % 6,
        "n_mod_12": n % 12,
        "n_mod_24": n % 24,
    }

    for s in range(12):
        row[f"beta_formula_shift_{s}"] = beta_formula_shift(n, s)
        row[f"matches_shift_{s}"] = (winner == beta_formula_shift(n, s))

    rows.append(row)

df = pd.DataFrame(rows)

# -----------------------------
# Switch points
# -----------------------------
switch_rows = []
prev = None
for _, row in df.iterrows():
    n = int(row["n"])
    w = int(row["winner_beta"])
    if prev is not None and w != prev:
        switch_rows.append({
            "switch_at_n": n,
            "new_winner_beta": w
        })
    prev = w

switch_df = pd.DataFrame(switch_rows)

# -----------------------------
# Run-length compression
# -----------------------------
runs = []
start_n = int(df.iloc[0]["n"])
current_winner = int(df.iloc[0]["winner_beta"])

for i in range(1, len(df)):
    n = int(df.iloc[i]["n"])
    w = int(df.iloc[i]["winner_beta"])
    if w != current_winner:
        end_n = int(df.iloc[i-1]["n"])
        runs.append({
            "start_n": start_n,
            "end_n": end_n,
            "winner_beta": current_winner,
            "length": end_n - start_n + 1
        })
        start_n = n
        current_winner = w

runs.append({
    "start_n": start_n,
    "end_n": int(df.iloc[-1]["n"]),
    "winner_beta": current_winner,
    "length": int(df.iloc[-1]["n"]) - start_n + 1
})

runs_df = pd.DataFrame(runs)

# -----------------------------
# Best simple 6-block law
# -----------------------------
shift_scores = []
for s in range(12):
    mismatches = int((~df[f"matches_shift_{s}"]).sum())
    shift_scores.append({
        "shift": s,
        "matches": int(df[f"matches_shift_{s}"].sum()),
        "mismatches": mismatches
    })

shift_scores_df = pd.DataFrame(shift_scores).sort_values(
    ["mismatches", "shift"]
).reset_index(drop=True)

best_shift = int(shift_scores_df.iloc[0]["shift"])
best_mismatch_df = df[~df[f"matches_shift_{best_shift}"]].copy()

# -----------------------------
# Residue summaries
# -----------------------------
mod12_summary = (
    df.groupby("n_mod_12", as_index=False)
      .agg(
          count=("n", "count"),
          beta_minus_wins=("winner_beta", lambda s: int(np.sum(s == -1))),
          beta_plus_wins=("winner_beta", lambda s: int(np.sum(s == 1))),
          ties=("winner_beta", lambda s: int(np.sum(s == 0))),
          avg_gap=("gap", "mean")
      )
      .sort_values("n_mod_12")
      .reset_index(drop=True)
)

mod24_summary = (
    df.groupby("n_mod_24", as_index=False)
      .agg(
          count=("n", "count"),
          beta_minus_wins=("winner_beta", lambda s: int(np.sum(s == -1))),
          beta_plus_wins=("winner_beta", lambda s: int(np.sum(s == 1))),
          ties=("winner_beta", lambda s: int(np.sum(s == 0))),
          avg_gap=("gap", "mean")
      )
      .sort_values("n_mod_24")
      .reset_index(drop=True)
)

# -----------------------------
# Summary prints
# -----------------------------
print(f"Verification range: n = 2,...,{N_MAX}")
print("All gaps equal 2:", bool((df["gap"] == 2).all()))
print("Number of switch points:", len(switch_df))
print("Best simple 6-block shift:", best_shift)
print("Mismatches for best simple 6-block shift:", int(shift_scores_df.iloc[0]["mismatches"]))

retro_table(shift_scores_df, "Fit of simple 6-block laws", max_rows=20)
retro_table(runs_df, "Compressed winner-beta runs", max_rows=40)
retro_table(switch_df, "Winner-beta switch points", max_rows=40)
retro_table(best_mismatch_df.head(80), f"Mismatches for best shift = {best_shift}", max_rows=80)
retro_table(mod12_summary, "Winner beta by n mod 12", max_rows=20)
retro_table(mod24_summary, "Winner beta by n mod 24", max_rows=30)

# -----------------------------
# Figures
# -----------------------------
# Figure 1: winner beta vs n
winner_plot = df.copy()
winner_plot["winner_label"] = winner_plot["winner_beta"].map({-1: "beta=-1", 1: "beta=+1", 0: "tie"})

fig1 = px.scatter(
    winner_plot,
    x="n",
    y="gap",
    color="winner_label",
    symbol="winner_label",
    title="Winning beta and gap"
)
apply_retro(fig1)
fig1.update_layout(xaxis_title="n", yaxis_title="gap")
fig1.show()

# Figure 2: run lengths
fig2 = px.bar(
    runs_df,
    x="start_n",
    y="length",
    color="winner_beta",
    text="end_n",
    title="Compressed winner-beta runs"
)
apply_retro(fig2)
fig2.update_traces(textposition="outside")
fig2.update_layout(xaxis_title="run start", yaxis_title="run length")
fig2.show()

# Figure 3: mod 12 summary
mod12_plot = mod12_summary.melt(
    id_vars=["n_mod_12", "count", "avg_gap"],
    value_vars=["beta_minus_wins", "beta_plus_wins", "ties"],
    var_name="category",
    value_name="value"
)
fig3 = px.bar(
    mod12_plot,
    x="n_mod_12",
    y="value",
    color="category",
    barmode="group",
    title="Winner counts by residue class mod 12"
)
apply_retro(fig3)
fig3.update_layout(xaxis_title="n mod 12", yaxis_title="count")
fig3.show()

# Figure 4: mod 24 summary
mod24_plot = mod24_summary.melt(
    id_vars=["n_mod_24", "count", "avg_gap"],
    value_vars=["beta_minus_wins", "beta_plus_wins", "ties"],
    var_name="category",
    value_name="value"
)
fig4 = px.bar(
    mod24_plot,
    x="n_mod_24",
    y="value",
    color="category",
    barmode="group",
    title="Winner counts by residue class mod 24"
)
apply_retro(fig4)
fig4.update_layout(xaxis_title="n mod 24", yaxis_title="count")
fig4.show()

# -----------------------------
# Save outputs
# -----------------------------
if SAVE_FILES:
    df.to_csv(os.path.join(OUTDIR, "week3g_full_winner_scan.csv"), index=False)
    runs_df.to_csv(os.path.join(OUTDIR, "week3g_runs.csv"), index=False)
    switch_df.to_csv(os.path.join(OUTDIR, "week3g_switch_points.csv"), index=False)
    shift_scores_df.to_csv(os.path.join(OUTDIR, "week3g_shift_scores.csv"), index=False)
    best_mismatch_df.to_csv(os.path.join(OUTDIR, "week3g_best_shift_mismatches.csv"), index=False)
    mod12_summary.to_csv(os.path.join(OUTDIR, "week3g_mod12_summary.csv"), index=False)
    mod24_summary.to_csv(os.path.join(OUTDIR, "week3g_mod24_summary.csv"), index=False)

    fig1.write_html(os.path.join(OUTDIR, "figure1_winner_gap.html"), include_plotlyjs="cdn")
    fig2.write_html(os.path.join(OUTDIR, "figure2_runs.html"), include_plotlyjs="cdn")
    fig3.write_html(os.path.join(OUTDIR, "figure3_mod12.html"), include_plotlyjs="cdn")
    fig4.write_html(os.path.join(OUTDIR, "figure4_mod24.html"), include_plotlyjs="cdn")

    print("\nSaved outputs in:")
    print(OUTDIR)
```

```python
# =========================================================
# WEEK 4 — Local interval geometry for H_n (CORRECTED)
# Focus:
#   Validate numerically that each partition interval I_j
#   contains exactly one interior critical point of H_n,
#   with derivative orientation adapted to the sign of H_n,
#   and that this critical point realizes the interval extremum.
#
# Family:
#   H_n(t) = 4 sin((2n-1)t/2) sin(t/2)
#
# Outputs:
#   1) summary table by n
#   2) detailed interval table
#   3) failure table (if any)
#   4) diagnostic plots for one chosen n
#   5) CSV exports for weekly zip
# =========================================================

import numpy as np
import pandas as pd
from math import pi
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MIN = 3
N_MAX = 160              # raise later if desired
GRID_PER_INTERVAL = 2500 # dense enough for robust bracketing
EDGE_EPS_FACTOR = 1e-4   # relative offset from endpoints for derivative sign checks
ROOT_TOL = 1e-10
MAX_BISECT_ITERS = 80
SAVE_CSV = True
MAKE_PLOTS = True
N_PLOT = 72              # representative case for figure

# Output paths (Colab-friendly)
OUT_DIR = "/content"

# -----------------------------
# Retro style helpers
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {"selector": "table", "props": [
                ("background-color", BG),
                ("color", TEXT),
                ("border-collapse", "collapse"),
                ("font-family", "Courier New, monospace")
            ]},
            {"selector": "th", "props": [
                ("background-color", "#111111"),
                ("color", NEON_AMBER),
                ("border", "1px solid #333333"),
                ("padding", "6px")
            ]},
            {"selector": "td", "props": [
                ("background-color", "#050505"),
                ("color", TEXT),
                ("border", "1px solid #222222"),
                ("padding", "6px")
            ]},
            {"selector": "caption", "props": [
                ("color", NEON_GREEN),
                ("font-size", "16px"),
                ("font-weight", "bold"),
                ("caption-side", "top"),
                ("padding", "8px")
            ]},
        ])
    )
    display(sty)

# -----------------------------
# Core functions
# -----------------------------
def H(n, t):
    return 4.0 * np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def Hprime(n, t):
    return (
        2.0*(2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
        + 2.0*np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0)
    )

def partition_interval(n, j):
    """
    I_j = (2π j/(2n-1), 2π (j+1)/(2n-1)), j=0,...,2n-2
    """
    a = 2.0*np.pi*j/(2*n - 1)
    b = 2.0*np.pi*(j+1)/(2*n - 1)
    return a, b

def interval_sign_expected(j):
    """
    For H_n on I_j, the sign alternates as (-1)^j:
    + on even j, - on odd j.
    """
    return 1 if (j % 2 == 0) else -1

def bisect_root(f, left, right, args=(), tol=ROOT_TOL, max_iter=MAX_BISECT_ITERS):
    fl = f(*args, left)
    fr = f(*args, right)
    if np.isnan(fl) or np.isnan(fr):
        return None
    if fl == 0:
        return left
    if fr == 0:
        return right
    if fl * fr > 0:
        return None

    a, b = left, right
    fa, fb = fl, fr
    for _ in range(max_iter):
        m = 0.5 * (a + b)
        fm = f(*args, m)
        if np.isnan(fm):
            return None
        if abs(fm) < tol or abs(b - a) < tol:
            return m
        if fa * fm <= 0:
            b, fb = m, fm
        else:
            a, fa = m, fm
    return 0.5 * (a + b)

def locate_roots_of_Hprime_in_interval(n, j, grid_per_interval=GRID_PER_INTERVAL):
    """
    Find all sign-change roots of H'_n inside I_j using dense bracketing + bisection.
    """
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = Hprime(n, x)

    roots = []

    # Catch near-zero samples
    near_zero_idx = np.where(np.abs(y) < 1e-8)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    # Catch sign changes
    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root(lambda nn, tt: Hprime(nn, tt), x[k], x[k+1], args=(n,))
            if r is not None:
                roots.append(float(r))

    # Deduplicate close roots
    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-7:
            dedup.append(r)

    # Keep only strictly interior roots
    interior = [r for r in dedup if (a < r < b)]
    return interior

def analyze_interval(n, j):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    sign_expected = interval_sign_expected(j)

    # endpoint values
    Ha = H(n, a)
    Hb = H(n, b)

    # derivative values near endpoints
    d_left = Hprime(n, a + eps)
    d_right = Hprime(n, b - eps)

    # critical points
    roots = locate_roots_of_Hprime_in_interval(n, j)
    num_crit = len(roots)
    crit = roots[0] if num_crit >= 1 else np.nan

    # dense numerical extremum of |H| for cross-check
    x_dense = np.linspace(a + eps, b - eps, GRID_PER_INTERVAL)
    H_dense = H(n, x_dense)
    absH_dense = np.abs(H_dense)

    idx_max_abs = int(np.argmax(absH_dense))
    t_max_abs = float(x_dense[idx_max_abs])
    H_at_max_abs = float(H_dense[idx_max_abs])
    absH_max = float(absH_dense[idx_max_abs])

    # if unique critical point, compare it with max of |H|
    Hcrit = float(H(n, crit)) if num_crit == 1 else np.nan
    absHcrit = abs(Hcrit) if num_crit == 1 else np.nan
    critical_matches_absmax = (
        bool(abs(crit - t_max_abs) < 2.0 * width / GRID_PER_INTERVAL)
        if num_crit == 1 else False
    )

    # sign consistency check for H on the interior
    probe = 0.5 * (a + b)
    H_probe = H(n, probe)
    sign_probe = 1 if H_probe > 0 else (-1 if H_probe < 0 else 0)
    sign_ok = (sign_probe == sign_expected)

    # orientation adapted to interval sign:
    # positive interval -> rise then fall
    # negative interval -> fall then rise
    if sign_expected == 1:
        left_orientation_ok = d_left > 0
        right_orientation_ok = d_right < 0
        extremum_type = "max"
    else:
        left_orientation_ok = d_left < 0
        right_orientation_ok = d_right > 0
        extremum_type = "min"

    uniqueness_ok = (num_crit == 1)
    orientation_ok = left_orientation_ok and right_orientation_ok
    interval_pass = uniqueness_ok and orientation_ok and sign_ok

    # extra classification
    if num_crit == 1:
        if Hcrit > 0:
            extremum_sign = "positive"
        elif Hcrit < 0:
            extremum_sign = "negative"
        else:
            extremum_sign = "zero"
    else:
        extremum_sign = "undetermined"

    return {
        "n": n,
        "j": j,
        "a": a,
        "b": b,
        "width": width,
        "sign_expected": sign_expected,
        "H(a)": Ha,
        "H(b)": Hb,
        "H(mid)": H_probe,
        "sign_ok": sign_ok,
        "d_left": d_left,
        "d_right": d_right,
        "left_orientation_ok": left_orientation_ok,
        "right_orientation_ok": right_orientation_ok,
        "orientation_ok": orientation_ok,
        "num_critical_points": num_crit,
        "critical_t": crit,
        "H_at_critical": Hcrit,
        "absH_at_critical": absHcrit,
        "t_max_abs_dense": t_max_abs,
        "H_at_max_abs_dense": H_at_max_abs,
        "absH_max_dense": absH_max,
        "critical_matches_absmax": critical_matches_absmax,
        "extremum_type_expected": extremum_type,
        "extremum_sign": extremum_sign,
        "interval_pass": interval_pass
    }

# -----------------------------
# Full scan
# -----------------------------
detail_rows = []

for n in range(N_MIN, N_MAX + 1):
    for j in range(0, 2*n - 1):   # j = 0,...,2n-2
        detail_rows.append(analyze_interval(n, j))

detail_df = pd.DataFrame(detail_rows)

# -----------------------------
# Summaries
# -----------------------------
summary_df = (
    detail_df.groupby("n", as_index=False)
    .agg(
        num_intervals=("j", "count"),
        all_sign_ok=("sign_ok", "all"),
        all_left_orientation_ok=("left_orientation_ok", "all"),
        all_right_orientation_ok=("right_orientation_ok", "all"),
        all_orientation_ok=("orientation_ok", "all"),
        all_unique_critical=("num_critical_points", lambda s: bool(np.all(s == 1))),
        all_match_absmax=("critical_matches_absmax", "all"),
        all_interval_pass=("interval_pass", "all"),
        min_absHmax=("absH_max_dense", "min"),
        max_absHmax=("absH_max_dense", "max"),
    )
)

fail_df = detail_df[~detail_df["interval_pass"]].copy()

# Additional diagnostic tables
nonunique_df = detail_df[detail_df["num_critical_points"] != 1].copy()
orientation_fail_df = detail_df[~detail_df["orientation_ok"]].copy()
sign_fail_df = detail_df[~detail_df["sign_ok"]].copy()

print("="*70)
print("WEEK 4 — LOCAL INTERVAL GEOMETRY SUMMARY (CORRECTED)")
print("="*70)
print(f"n range scanned: {N_MIN} to {N_MAX}")
print(f"Total intervals checked: {len(detail_df)}")
print(f"n values with all intervals passing: {int(summary_df['all_interval_pass'].sum())} / {len(summary_df)}")
print(f"Total failing intervals: {len(fail_df)}")
print(f"Intervals with num_critical_points != 1: {len(nonunique_df)}")
print(f"Intervals with orientation failure: {len(orientation_fail_df)}")
print(f"Intervals with sign mismatch failure: {len(sign_fail_df)}")

retro_table(summary_df, "Week 4 summary by n (corrected)", max_rows=25)

if len(fail_df) == 0:
    print("\nNo interval-level failures detected in the scanned range.")
else:
    retro_table(
        fail_df[[
            "n", "j", "sign_expected", "num_critical_points",
            "left_orientation_ok", "right_orientation_ok", "orientation_ok",
            "sign_ok", "critical_t", "t_max_abs_dense", "interval_pass"
        ]],
        "Interval-level failures (corrected)",
        max_rows=30
    )

# Optional compact tables
retro_table(
    detail_df[[
        "n", "j", "sign_expected", "num_critical_points",
        "left_orientation_ok", "right_orientation_ok", "orientation_ok",
        "critical_t", "absH_at_critical", "absH_max_dense",
        "critical_matches_absmax", "extremum_type_expected", "interval_pass"
    ]],
    "First interval records (corrected)",
    max_rows=20
)

# -----------------------------
# Plot 1: pass/fail heat-style map
# -----------------------------
if MAKE_PLOTS:
    plot_df = detail_df.copy()
    plot_df["pass_int"] = plot_df["interval_pass"].astype(int)

    fig1 = px.scatter(
        plot_df,
        x="n",
        y="j",
        color="pass_int",
        hover_data=[
            "num_critical_points", "orientation_ok", "sign_ok",
            "critical_matches_absmax", "extremum_type_expected"
        ],
        title="Week 4 interval check map (corrected): pass/fail by (n,j)"
    )
    apply_retro(fig1)
    fig1.update_layout(xaxis_title="n", yaxis_title="interval index j")
    fig1.show()

# -----------------------------
# Plot 2: diagnostic profile for a representative n
# -----------------------------
if MAKE_PLOTS and (N_PLOT >= N_MIN) and (N_PLOT <= N_MAX):
    n = N_PLOT
    t = np.linspace(0, 2*np.pi, 14000)
    y = H(n, t)
    yp = Hprime(n, t)

    fig2 = go.Figure()
    fig2.add_trace(go.Scatter(
        x=t, y=y,
        mode="lines",
        name=f"H_{n}(t)",
        line=dict(width=2, color=NEON_AMBER)
    ))
    fig2.add_trace(go.Scatter(
        x=[0, 2*np.pi], y=[0, 0],
        mode="lines",
        name="y=0",
        line=dict(width=1, dash="dash", color=NEON_GREEN)
    ))

    # Partition points
    partition_pts = [2*np.pi*j/(2*n - 1) for j in range(2*n)]
    fig2.add_trace(go.Scatter(
        x=partition_pts,
        y=[0.0]*len(partition_pts),
        mode="markers",
        name="partition zeros",
        marker=dict(size=5, color=NEON_CYAN)
    ))

    # Critical points
    crit_x = []
    crit_y = []
    for j in range(0, 2*n - 1):
        roots = locate_roots_of_Hprime_in_interval(n, j)
        if len(roots) == 1:
            r = roots[0]
            crit_x.append(r)
            crit_y.append(H(n, r))

    fig2.add_trace(go.Scatter(
        x=crit_x,
        y=crit_y,
        mode="markers",
        name="unique critical points",
        marker=dict(size=8, color=NEON_PINK)
    ))

    apply_retro(fig2, f"Week 4 diagnostic profile for H_{n}(t) (corrected)")
    fig2.update_layout(xaxis_title="t", yaxis_title="value")
    fig2.show()

    # Derivative profile
    fig3 = go.Figure()
    fig3.add_trace(go.Scatter(
        x=t, y=yp,
        mode="lines",
        name=f"H_{n}'(t)",
        line=dict(width=2, color=NEON_CYAN)
    ))
    fig3.add_trace(go.Scatter(
        x=[0, 2*np.pi], y=[0, 0],
        mode="lines",
        name="y=0",
        line=dict(width=1, dash="dash", color=NEON_GREEN)
    ))
    apply_retro(fig3, f"Derivative profile for H_{n}'(t) (corrected)")
    fig3.update_layout(xaxis_title="t", yaxis_title="value")
    fig3.show()

# -----------------------------
# CSV export
# -----------------------------
if SAVE_CSV:
    summary_path = f"{OUT_DIR}/week4_interval_geometry_summary_corrected.csv"
    detail_path = f"{OUT_DIR}/week4_interval_geometry_detail_corrected.csv"
    fail_path = f"{OUT_DIR}/week4_interval_geometry_failures_corrected.csv"

    summary_df.to_csv(summary_path, index=False)
    detail_df.to_csv(detail_path, index=False)
    fail_df.to_csv(fail_path, index=False)

    print("\nSaved CSV files:")
    print(summary_path)
    print(detail_path)
    print(fail_path)

# -----------------------------
# Suggested manuscript sentence
# -----------------------------
if len(fail_df) == 0:
    print(
        f"\nSuggested numerical statement:\n"
        f'For all n in [{N_MIN}, {N_MAX}], every partition interval I_j contained exactly one '
        f'interior critical point of H_n. Moreover, on intervals where H_n is positive, the '
        f'derivative was positive near the left endpoint and negative near the right endpoint, '
        f'while on intervals where H_n is negative, the derivative was negative near the left '
        f'endpoint and positive near the right endpoint. In all tested cases, the critical point '
        f'agreed with the numerical interval extremum.'
    )
else:
    print(
        "\nAt least one interval-level failure was detected. Inspect fail_df before writing the manuscript claim."
    )
```

```python
# =========================================================
# WEEK 5 — Local interval counting rule
# Focus:
#   Validate the 0/1/2 local counting rule for H_n(t)=beta
#   interval by interval.
#
# Family:
#   H_n(t) = 4 sin((2n-1)t/2) sin(t/2)
#
# Target rule:
#   On each partition interval I_j:
#   - if sign(H_n on I_j) != beta: contribution = 0
#   - else compare M_{n,j} = max_{I_j} |H_n|
#       M_{n,j} < 1  -> 0 solutions
#       M_{n,j} = 1  -> 1 solution
#       M_{n,j} > 1  -> 2 solutions
#
# Outputs:
#   1) summary table by (n,beta)
#   2) detailed interval table
#   3) failure table (if any)
#   4) diagnostic plots for one chosen n and beta
#   5) CSV exports for weekly zip
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MIN = 3
N_MAX = 160
BETAS = [-1, 1]

GRID_PER_INTERVAL = 4000
EDGE_EPS_FACTOR = 1e-4
ROOT_TOL = 1e-10
LEVEL_TOL = 1e-8
MAX_BISECT_ITERS = 80

SAVE_CSV = True
MAKE_PLOTS = True
N_PLOT = 72
BETA_PLOT = 1

OUT_DIR = "/content"

# -----------------------------
# Retro style helpers
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {"selector": "table", "props": [
                ("background-color", BG),
                ("color", TEXT),
                ("border-collapse", "collapse"),
                ("font-family", "Courier New, monospace")
            ]},
            {"selector": "th", "props": [
                ("background-color", "#111111"),
                ("color", NEON_AMBER),
                ("border", "1px solid #333333"),
                ("padding", "6px")
            ]},
            {"selector": "td", "props": [
                ("background-color", "#050505"),
                ("color", TEXT),
                ("border", "1px solid #222222"),
                ("padding", "6px")
            ]},
            {"selector": "caption", "props": [
                ("color", NEON_GREEN),
                ("font-size", "16px"),
                ("font-weight", "bold"),
                ("caption-side", "top"),
                ("padding", "8px")
            ]},
        ])
    )
    display(sty)

# -----------------------------
# Core functions
# -----------------------------
def H(n, t):
    return 4.0 * np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def Hprime(n, t):
    return (
        2.0*(2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
        + 2.0*np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0)
    )

def partition_interval(n, j):
    a = 2.0*np.pi*j/(2*n - 1)
    b = 2.0*np.pi*(j+1)/(2*n - 1)
    return a, b

def interval_sign_expected(j):
    return 1 if (j % 2 == 0) else -1

def bisect_root_scalar(f, left, right, tol=ROOT_TOL, max_iter=MAX_BISECT_ITERS):
    fl = f(left)
    fr = f(right)
    if np.isnan(fl) or np.isnan(fr):
        return None
    if abs(fl) < tol:
        return left
    if abs(fr) < tol:
        return right
    if fl * fr > 0:
        return None

    a, b = left, right
    fa, fb = fl, fr
    for _ in range(max_iter):
        m = 0.5 * (a + b)
        fm = f(m)
        if np.isnan(fm):
            return None
        if abs(fm) < tol or abs(b - a) < tol:
            return m
        if fa * fm <= 0:
            b, fb = m, fm
        else:
            a, fa = m, fm
    return 0.5 * (a + b)

def locate_roots_of_Hprime_in_interval(n, j, grid_per_interval=GRID_PER_INTERVAL):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = Hprime(n, x)

    roots = []

    near_zero_idx = np.where(np.abs(y) < 1e-8)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(lambda tt: Hprime(n, tt), x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-7:
            dedup.append(r)

    return [r for r in dedup if (a < r < b)]

def locate_roots_of_level_in_interval(n, j, beta, grid_per_interval=GRID_PER_INTERVAL):
    """
    Count solutions of H_n(t)=beta inside I_j.
    """
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    f = lambda tt: H(n, tt) - beta

    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = f(x)

    roots = []

    near_zero_idx = np.where(np.abs(y) < 1e-7)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(f, x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-6:
            dedup.append(r)

    return [r for r in dedup if (a < r < b)]

def analyze_interval_beta(n, j, beta):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    sign_expected = interval_sign_expected(j)
    sign_matches_beta = (sign_expected == beta)

    # critical point data (imported from Week 4 logic)
    crits = locate_roots_of_Hprime_in_interval(n, j)
    num_crit = len(crits)
    crit = crits[0] if num_crit == 1 else np.nan

    # local maximum of |H|
    if num_crit == 1:
        Hcrit = float(H(n, crit))
        M = abs(Hcrit)
    else:
        # fallback dense estimate, though Week 4 suggests this should not happen
        x_dense = np.linspace(a + eps, b - eps, GRID_PER_INTERVAL)
        M = float(np.max(np.abs(H(n, x_dense))))
        Hcrit = np.nan

    # real roots of H_n(t)=beta in the interval
    level_roots = locate_roots_of_level_in_interval(n, j, beta)
    actual_count = len(level_roots)

    # predicted count
    if not sign_matches_beta:
        predicted_count = 0
        regime = "sign-mismatch"
    else:
        if M < 1.0 - LEVEL_TOL:
            predicted_count = 0
            regime = "below"
        elif M > 1.0 + LEVEL_TOL:
            predicted_count = 2
            regime = "above"
        else:
            predicted_count = 1
            regime = "tangent"

    rule_ok = (predicted_count == actual_count)

    return {
        "n": n,
        "j": j,
        "beta": beta,
        "sign_expected": sign_expected,
        "sign_matches_beta": sign_matches_beta,
        "num_critical_points": num_crit,
        "critical_t": crit,
        "H_at_critical": Hcrit,
        "M_nj": M,
        "regime": regime,
        "predicted_count": predicted_count,
        "actual_count": actual_count,
        "rule_ok": rule_ok,
        "root_1": level_roots[0] if len(level_roots) >= 1 else np.nan,
        "root_2": level_roots[1] if len(level_roots) >= 2 else np.nan,
    }

# -----------------------------
# Full scan
# -----------------------------
detail_rows = []

for n in range(N_MIN, N_MAX + 1):
    for beta in BETAS:
        for j in range(0, 2*n - 1):
            detail_rows.append(analyze_interval_beta(n, j, beta))

detail_df = pd.DataFrame(detail_rows)

# -----------------------------
# Summaries
# -----------------------------
summary_df = (
    detail_df.groupby(["n", "beta"], as_index=False)
    .agg(
        num_intervals=("j", "count"),
        all_unique_critical=("num_critical_points", lambda s: bool(np.all(s == 1))),
        all_rule_ok=("rule_ok", "all"),
        num_rule_failures=("rule_ok", lambda s: int((~s).sum())),
        num_sign_match_intervals=("sign_matches_beta", "sum"),
        num_predicted_0=("predicted_count", lambda s: int(np.sum(s == 0))),
        num_predicted_1=("predicted_count", lambda s: int(np.sum(s == 1))),
        num_predicted_2=("predicted_count", lambda s: int(np.sum(s == 2))),
        num_actual_0=("actual_count", lambda s: int(np.sum(s == 0))),
        num_actual_1=("actual_count", lambda s: int(np.sum(s == 1))),
        num_actual_2=("actual_count", lambda s: int(np.sum(s == 2))),
    )
)

fail_df = detail_df[~detail_df["rule_ok"]].copy()

regime_summary_df = (
    detail_df.groupby(["beta", "regime"], as_index=False)
    .size()
    .rename(columns={"size": "count"})
    .sort_values(["beta", "regime"])
    .reset_index(drop=True)
)

print("="*72)
print("WEEK 5 — LOCAL INTERVAL COUNTING RULE SUMMARY")
print("="*72)
print(f"n range scanned: {N_MIN} to {N_MAX}")
print(f"beta values scanned: {BETAS}")
print(f"Total interval-beta cases checked: {len(detail_df)}")
print(f"Total failures of local 0/1/2 rule: {len(fail_df)}")

retro_table(summary_df, "Week 5 summary by (n,beta)", max_rows=30)
retro_table(regime_summary_df, "Week 5 regime summary", max_rows=20)

if len(fail_df) == 0:
    print("\nNo failures detected for the local 0/1/2 interval counting rule in the scanned range.")
else:
    retro_table(
        fail_df[[
            "n", "j", "beta", "sign_expected", "sign_matches_beta",
            "M_nj", "regime", "predicted_count", "actual_count",
            "critical_t", "root_1", "root_2"
        ]],
        "Failures of local 0/1/2 rule",
        max_rows=30
    )

retro_table(
    detail_df[[
        "n", "j", "beta", "sign_expected", "sign_matches_beta",
        "M_nj", "regime", "predicted_count", "actual_count", "rule_ok"
    ]],
    "First interval-beta records",
    max_rows=20
)

# -----------------------------
# Plot 1: pass/fail map
# -----------------------------
if MAKE_PLOTS:
    plot_df = detail_df.copy()
    plot_df["rule_ok_int"] = plot_df["rule_ok"].astype(int)
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig1 = px.scatter(
        plot_df,
        x="n",
        y="j",
        color="rule_ok_int",
        symbol="beta_str",
        hover_data=[
            "beta", "sign_matches_beta", "M_nj",
            "regime", "predicted_count", "actual_count"
        ],
        title="Week 5 map: local 0/1/2 counting rule by (n,j,beta)"
    )
    apply_retro(fig1)
    fig1.update_layout(xaxis_title="n", yaxis_title="interval index j")
    fig1.show()

# -----------------------------
# Plot 2: diagnostic profile for chosen (n,beta)
# -----------------------------
if MAKE_PLOTS and (N_PLOT >= N_MIN) and (N_PLOT <= N_MAX) and (BETA_PLOT in BETAS):
    n = N_PLOT
    beta = BETA_PLOT
    t = np.linspace(0, 2*np.pi, 20000)
    y = H(n, t)

    fig2 = go.Figure()
    fig2.add_trace(go.Scatter(
        x=t, y=y,
        mode="lines",
        name=f"H_{n}(t)",
        line=dict(width=2, color=NEON_AMBER)
    ))
    fig2.add_trace(go.Scatter(
        x=[0, 2*np.pi], y=[beta, beta],
        mode="lines",
        name=f"y={beta}",
        line=dict(width=1.5, dash="dash", color=NEON_GREEN)
    ))
    fig2.add_trace(go.Scatter(
        x=[0, 2*np.pi], y=[0, 0],
        mode="lines",
        name="y=0",
        line=dict(width=1, dash="dot", color=NEON_CYAN)
    ))

    # Partition points
    partition_pts = [2*np.pi*j/(2*n - 1) for j in range(2*n)]
    fig2.add_trace(go.Scatter(
        x=partition_pts,
        y=[0.0]*len(partition_pts),
        mode="markers",
        name="partition zeros",
        marker=dict(size=4, color=NEON_CYAN)
    ))

    # Level-crossing roots
    root_x = []
    root_y = []
    for j in range(0, 2*n - 1):
        roots = locate_roots_of_level_in_interval(n, j, beta)
        for r in roots:
            root_x.append(r)
            root_y.append(beta)

    fig2.add_trace(go.Scatter(
        x=root_x,
        y=root_y,
        mode="markers",
        name="solutions of H_n(t)=beta",
        marker=dict(size=7, color=NEON_PINK)
    ))

    apply_retro(fig2, f"Week 5 diagnostic profile for H_{n}(t) with beta={beta}")
    fig2.update_layout(xaxis_title="t", yaxis_title="value")
    fig2.show()

# -----------------------------
# Plot 3: M_nj distribution vs threshold 1
# -----------------------------
if MAKE_PLOTS:
    mm = detail_df[detail_df["sign_matches_beta"]].copy()
    mm["beta_str"] = mm["beta"].astype(str)

    fig3 = px.scatter(
        mm,
        x="n",
        y="M_nj",
        color="beta_str",
        hover_data=["j", "regime", "predicted_count", "actual_count"],
        title="Week 5: local extrema M_{n,j} on sign-compatible intervals"
    )
    apply_retro(fig3)
    fig3.add_trace(go.Scatter(
        x=[mm["n"].min(), mm["n"].max()],
        y=[1, 1],
        mode="lines",
        name="threshold 1",
        line=dict(width=1.5, dash="dash", color=NEON_GREEN)
    ))
    fig3.update_layout(xaxis_title="n", yaxis_title="M_{n,j}")
    fig3.show()

# -----------------------------
# CSV export
# -----------------------------
if SAVE_CSV:
    summary_path = f"{OUT_DIR}/week5_local_counting_summary.csv"
    detail_path = f"{OUT_DIR}/week5_local_counting_detail.csv"
    fail_path = f"{OUT_DIR}/week5_local_counting_failures.csv"
    regime_path = f"{OUT_DIR}/week5_local_counting_regime_summary.csv"

    summary_df.to_csv(summary_path, index=False)
    detail_df.to_csv(detail_path, index=False)
    fail_df.to_csv(fail_path, index=False)
    regime_summary_df.to_csv(regime_path, index=False)

    print("\nSaved CSV files:")
    print(summary_path)
    print(detail_path)
    print(fail_path)
    print(regime_path)

# -----------------------------
# Suggested manuscript sentence
# -----------------------------
if len(fail_df) == 0:
    print(
        f"\nSuggested numerical statement:\n"
        f'For all n in [{N_MIN}, {N_MAX}] and both beta=±1, the interval-by-interval '
        f'0/1/2 counting rule matched the exact number of solutions of H_n(t)=beta in every '
        f'partition interval. In particular, sign-incompatible intervals contributed no solutions, '
        f'whereas sign-compatible intervals contributed 0, 1, or 2 solutions according as '
        f'the local extremum M_(n,j) lay below, at, or above the threshold level 1.'
    )
else:
    print(
        "\nAt least one failure of the local 0/1/2 rule was detected. Inspect fail_df before writing the manuscript claim."
    )
```

```python
# =========================================================
# WEEK 6 — Global interval decomposition vs polynomial count
# Focus:
#   Validate that
#       U(Q_{n,n-1}^{-1,beta}) = sum_j kappa_{n,j}^{(beta)}
#   where kappa_{n,j}^{(beta)} is the local interval contribution.
#
# Family:
#   Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
#   beta in {-1, 1}
#
# Outputs:
#   1) summary table by (n,beta)
#   2) failure table (if any)
#   3) comparison plots
#   4) CSV exports for weekly zip
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MIN = 3
N_MAX = 160
BETAS = [-1, 1]

UNIT_TOL = 1e-7
GRID_PER_INTERVAL = 4000
EDGE_EPS_FACTOR = 1e-4
ROOT_TOL = 1e-10
LEVEL_TOL = 1e-8
MAX_BISECT_ITERS = 80

SAVE_CSV = True
MAKE_PLOTS = True

OUT_DIR = "/content"

# -----------------------------
# Retro style helpers
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {"selector": "table", "props": [
                ("background-color", BG),
                ("color", TEXT),
                ("border-collapse", "collapse"),
                ("font-family", "Courier New, monospace")
            ]},
            {"selector": "th", "props": [
                ("background-color", "#111111"),
                ("color", NEON_AMBER),
                ("border", "1px solid #333333"),
                ("padding", "6px")
            ]},
            {"selector": "td", "props": [
                ("background-color", "#050505"),
                ("color", TEXT),
                ("border", "1px solid #222222"),
                ("padding", "6px")
            ]},
            {"selector": "caption", "props": [
                ("color", NEON_GREEN),
                ("font-size", "16px"),
                ("font-weight", "bold"),
                ("caption-side", "top"),
                ("padding", "8px")
            ]},
        ])
    )
    display(sty)

# -----------------------------
# Core functions
# -----------------------------
def H(n, t):
    return 4.0 * np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def Hprime(n, t):
    return (
        2.0*(2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
        + 2.0*np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0)
    )

def partition_interval(n, j):
    a = 2.0*np.pi*j/(2*n - 1)
    b = 2.0*np.pi*(j+1)/(2*n - 1)
    return a, b

def interval_sign_expected(j):
    return 1 if (j % 2 == 0) else -1

def bisect_root_scalar(f, left, right, tol=ROOT_TOL, max_iter=MAX_BISECT_ITERS):
    fl = f(left)
    fr = f(right)
    if np.isnan(fl) or np.isnan(fr):
        return None
    if abs(fl) < tol:
        return left
    if abs(fr) < tol:
        return right
    if fl * fr > 0:
        return None

    a, b = left, right
    fa, fb = fl, fr
    for _ in range(max_iter):
        m = 0.5 * (a + b)
        fm = f(m)
        if np.isnan(fm):
            return None
        if abs(fm) < tol or abs(b - a) < tol:
            return m
        if fa * fm <= 0:
            b, fb = m, fm
        else:
            a, fa = m, fm
    return 0.5 * (a + b)

def locate_roots_of_Hprime_in_interval(n, j, grid_per_interval=GRID_PER_INTERVAL):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = Hprime(n, x)

    roots = []

    near_zero_idx = np.where(np.abs(y) < 1e-8)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(lambda tt: Hprime(n, tt), x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-7:
            dedup.append(r)

    return [r for r in dedup if (a < r < b)]

def locate_roots_of_level_in_interval(n, j, beta, grid_per_interval=GRID_PER_INTERVAL):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width

    f = lambda tt: H(n, tt) - beta

    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = f(x)

    roots = []

    near_zero_idx = np.where(np.abs(y) < 1e-7)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(f, x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-6:
            dedup.append(r)

    return [r for r in dedup if (a < r < b)]

# -----------------------------
# Polynomial side
# -----------------------------
def q_coeffs_edge(n, beta):
    """
    Q_{n,n-1}^{-1,beta}(z) = z^(2n) - z^(2n-1) + beta z^n - z + 1
    """
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0           # z^(2n)
    coeffs[1] = -1.0          # z^(2n-1)
    coeffs[n] = float(beta)   # z^n
    coeffs[-2] = -1.0         # z
    coeffs[-1] = 1.0          # constant
    return coeffs

def q_roots_edge(n, beta):
    return np.roots(q_coeffs_edge(n, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

# -----------------------------
# Local contribution kappa
# -----------------------------
def local_kappa(n, j, beta):
    """
    Compute kappa_{n,j}^{(beta)} = # { t in I_j : H_n(t)=beta }.
    """
    roots = locate_roots_of_level_in_interval(n, j, beta)
    return len(roots)

# -----------------------------
# Main scan
# -----------------------------
interval_rows = []
global_rows = []

for n in range(N_MIN, N_MAX + 1):
    for beta in BETAS:
        kappas = []
        for j in range(0, 2*n - 1):
            kappa = local_kappa(n, j, beta)
            kappas.append(kappa)
            interval_rows.append({
                "n": n,
                "beta": beta,
                "j": j,
                "kappa_nj": kappa,
                "interval_sign": interval_sign_expected(j)
            })

        local_sum = int(np.sum(kappas))

        roots = q_roots_edge(n, beta)
        U_poly = count_unimodular_roots(roots)

        global_rows.append({
            "n": n,
            "beta": beta,
            "U_polynomial": U_poly,
            "U_interval_sum": local_sum,
            "match": (U_poly == local_sum),
            "difference": U_poly - local_sum,
            "num_intervals": 2*n - 1,
            "num_kappa_0": int(np.sum(np.array(kappas) == 0)),
            "num_kappa_1": int(np.sum(np.array(kappas) == 1)),
            "num_kappa_2": int(np.sum(np.array(kappas) == 2)),
        })

interval_df = pd.DataFrame(interval_rows)
global_df = pd.DataFrame(global_rows)

fail_df = global_df[~global_df["match"]].copy()

# -----------------------------
# Summary print
# -----------------------------
print("="*72)
print("WEEK 6 — GLOBAL INTERVAL DECOMPOSITION SUMMARY")
print("="*72)
print(f"n range scanned: {N_MIN} to {N_MAX}")
print(f"beta values scanned: {BETAS}")
print(f"Total (n,beta) cases checked: {len(global_df)}")
print(f"Total failures of global decomposition: {len(fail_df)}")

retro_table(global_df, "Week 6 global summary by (n,beta)", max_rows=40)

if len(fail_df) == 0:
    print("\nNo failures detected: U_polynomial = U_interval_sum in the scanned range.")
else:
    retro_table(
        fail_df[[
            "n", "beta", "U_polynomial", "U_interval_sum",
            "difference", "num_kappa_0", "num_kappa_1", "num_kappa_2"
        ]],
        "Failures of global interval decomposition",
        max_rows=30
    )

retro_table(
    interval_df[["n", "beta", "j", "kappa_nj", "interval_sign"]],
    "First interval contribution records",
    max_rows=20
)

# -----------------------------
# Plot 1: polynomial count vs interval sum
# -----------------------------
if MAKE_PLOTS:
    plot_df = global_df.copy()
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig1 = px.scatter(
        plot_df,
        x="U_polynomial",
        y="U_interval_sum",
        color="beta_str",
        hover_data=["n", "difference", "num_kappa_0", "num_kappa_1", "num_kappa_2"],
        title="Week 6: polynomial count vs global interval sum"
    )
    apply_retro(fig1)
    minv = min(plot_df["U_polynomial"].min(), plot_df["U_interval_sum"].min())
    maxv = max(plot_df["U_polynomial"].max(), plot_df["U_interval_sum"].max())
    fig1.add_trace(go.Scatter(
        x=[minv, maxv],
        y=[minv, maxv],
        mode="lines",
        name="y=x",
        line=dict(width=1.5, dash="dash", color=NEON_GREEN)
    ))
    fig1.update_layout(xaxis_title="U_polynomial", yaxis_title="U_interval_sum")
    fig1.show()

# -----------------------------
# Plot 2: counts by n
# -----------------------------
if MAKE_PLOTS:
    plot_df = global_df.copy()
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig2 = px.line(
        plot_df,
        x="n",
        y="U_polynomial",
        color="beta_str",
        markers=True,
        title="Week 6: total unimodular-root count by n"
    )
    apply_retro(fig2)
    fig2.update_layout(xaxis_title="n", yaxis_title="U(Q)")
    fig2.show()

# -----------------------------
# Plot 3: decomposition pattern
# -----------------------------
if MAKE_PLOTS:
    plot_df = global_df.copy()
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig3 = go.Figure()
    for beta in BETAS:
        sub = plot_df[plot_df["beta"] == beta]
        fig3.add_trace(go.Scatter(
            x=sub["n"], y=sub["num_kappa_2"],
            mode="lines+markers",
            name=f"num intervals with kappa=2 (beta={beta})"
        ))
        fig3.add_trace(go.Scatter(
            x=sub["n"], y=sub["num_kappa_1"],
            mode="lines+markers",
            name=f"num intervals with kappa=1 (beta={beta})"
        ))
    apply_retro(fig3, "Week 6: decomposition of global count by interval contributions")
    fig3.update_layout(xaxis_title="n", yaxis_title="number of intervals")
    fig3.show()

# -----------------------------
# CSV export
# -----------------------------
if SAVE_CSV:
    global_path = f"{OUT_DIR}/week6_global_interval_summary.csv"
    interval_path = f"{OUT_DIR}/week6_global_interval_detail.csv"
    fail_path = f"{OUT_DIR}/week6_global_interval_failures.csv"

    global_df.to_csv(global_path, index=False)
    interval_df.to_csv(interval_path, index=False)
    fail_df.to_csv(fail_path, index=False)

    print("\nSaved CSV files:")
    print(global_path)
    print(interval_path)
    print(fail_path)

# -----------------------------
# Suggested manuscript sentence
# -----------------------------
if len(fail_df) == 0:
    print(
        f"\nSuggested numerical statement:\n"
        f'For all n in [{N_MIN}, {N_MAX}] and both beta=±1, the total number of unimodular zeros '
        f'of Q_(n,n-1)^(-1,beta) matched exactly the sum of the local interval contributions '
        f'kappa_(n,j)^(beta). Thus the global count agrees perfectly with the interval-by-interval '
        f'decomposition throughout the tested range.'
    )
else:
    print(
        "\nAt least one failure of the global interval decomposition was detected. "
        "Inspect fail_df before writing the manuscript claim."
    )
```

```python
# =========================================================
# WEEK 7 CLOSEOUT — Integrated summary of Weeks 4–7
# Focus:
#   Produce one compact computational summary showing:
#   (i) unique critical point per interval,
#   (ii) exact local 0/1/2 counting rule,
#   (iii) exact global interval decomposition vs polynomial count.
#
# This is NOT a new proof block.
# It is a closeout / synthesis block for the weekly package.
# =========================================================

import numpy as np
import pandas as pd
from IPython.display import display
import plotly.express as px
import plotly.graph_objects as go

# -----------------------------
# Parameters
# -----------------------------
N_MIN = 3
N_MAX = 160
BETAS = [-1, 1]

UNIT_TOL = 1e-7
GRID_PER_INTERVAL = 2000
EDGE_EPS_FACTOR = 1e-4
ROOT_TOL = 1e-10
LEVEL_TOL = 1e-8
MAX_BISECT_ITERS = 80

SAVE_CSV = True
MAKE_PLOTS = True
N_PLOT = 21
BETA_PLOT = 1

OUT_DIR = "/content"

# -----------------------------
# Styling
# -----------------------------
BG = "#000000"
PAPER = "#000000"
GRID = "rgba(0,255,180,0.14)"
TEXT = "#F5E6A8"
NEON_GREEN = "#39FF14"
NEON_CYAN = "#00F5D4"
NEON_AMBER = "#FFB000"
NEON_PINK = "#FF4FD8"

def apply_retro(fig, title=None):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor=PAPER,
        plot_bgcolor=BG,
        font=dict(family="Courier New, monospace", color=TEXT, size=14),
        title=dict(
            text=title or "",
            x=0.5,
            xanchor="center",
            font=dict(size=20, color=NEON_AMBER)
        ),
        xaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        yaxis=dict(
            showgrid=True,
            gridcolor=GRID,
            zeroline=True,
            zerolinecolor="rgba(255,176,0,0.25)",
            showline=True,
            linecolor="rgba(245,230,168,0.35)"
        ),
        legend=dict(
            bgcolor="rgba(0,0,0,0.35)",
            bordercolor="rgba(57,255,20,0.25)",
            borderwidth=1,
            font=dict(color=TEXT)
        ),
        margin=dict(l=50, r=30, t=70, b=50),
    )
    return fig

def retro_table(df, caption=None, max_rows=20):
    show = df.head(max_rows).copy()
    sty = (
        show.style
        .set_caption(caption if caption else "")
        .set_table_styles([
            {"selector": "table", "props": [
                ("background-color", BG),
                ("color", TEXT),
                ("border-collapse", "collapse"),
                ("font-family", "Courier New, monospace")
            ]},
            {"selector": "th", "props": [
                ("background-color", "#111111"),
                ("color", NEON_AMBER),
                ("border", "1px solid #333333"),
                ("padding", "6px")
            ]},
            {"selector": "td", "props": [
                ("background-color", "#050505"),
                ("color", TEXT),
                ("border", "1px solid #222222"),
                ("padding", "6px")
            ]},
            {"selector": "caption", "props": [
                ("color", NEON_GREEN),
                ("font-size", "16px"),
                ("font-weight", "bold"),
                ("caption-side", "top"),
                ("padding", "8px")
            ]},
        ])
    )
    display(sty)

# -----------------------------
# Core functions
# -----------------------------
def partition_interval(n, j):
    a = 2.0*np.pi*j/(2*n - 1)
    b = 2.0*np.pi*(j+1)/(2*n - 1)
    return a, b

def interval_sign_expected(j):
    return 1 if (j % 2 == 0) else -1

def H(n, t):
    return 4.0 * np.sin((2*n - 1)*t/2.0) * np.sin(t/2.0)

def Hprime(n, t):
    return (
        2.0*(2*n - 1)*np.cos((2*n - 1)*t/2.0)*np.sin(t/2.0)
        + 2.0*np.sin((2*n - 1)*t/2.0)*np.cos(t/2.0)
    )

def F(n, t):
    return (2*n - 1)/np.tan((2*n - 1)*t/2.0) + 1.0/np.tan(t/2.0)

def Fprime(n, t):
    return (
        -((2*n - 1)**2)/2.0 * (1.0 / (np.sin((2*n - 1)*t/2.0)**2))
        -0.5 * (1.0 / (np.sin(t/2.0)**2))
    )

def bisect_root_scalar(f, left, right, tol=ROOT_TOL, max_iter=MAX_BISECT_ITERS):
    fl = f(left)
    fr = f(right)
    if np.isnan(fl) or np.isnan(fr):
        return None
    if abs(fl) < tol:
        return left
    if abs(fr) < tol:
        return right
    if fl * fr > 0:
        return None

    a, b = left, right
    fa, fb = fl, fr
    for _ in range(max_iter):
        m = 0.5 * (a + b)
        fm = f(m)
        if np.isnan(fm):
            return None
        if abs(fm) < tol or abs(b - a) < tol:
            return m
        if fa * fm <= 0:
            b, fb = m, fm
        else:
            a, fa = m, fm
    return 0.5 * (a + b)

def locate_roots_of_F_in_interval(n, j, grid_per_interval=GRID_PER_INTERVAL):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width
    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = F(n, x)

    roots = []
    near_zero_idx = np.where(np.abs(y) < 1e-8)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(lambda tt: F(n, tt), x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-7:
            dedup.append(r)
    return [r for r in dedup if (a < r < b)]

def locate_roots_of_level_in_interval(n, j, beta, grid_per_interval=GRID_PER_INTERVAL):
    a, b = partition_interval(n, j)
    width = b - a
    eps = EDGE_EPS_FACTOR * width
    f = lambda tt: H(n, tt) - beta
    x = np.linspace(a + eps, b - eps, grid_per_interval)
    y = f(x)

    roots = []
    near_zero_idx = np.where(np.abs(y) < 1e-7)[0]
    for idx in near_zero_idx:
        roots.append(float(x[idx]))

    for k in range(len(x) - 1):
        y1, y2 = y[k], y[k+1]
        if np.isnan(y1) or np.isnan(y2):
            continue
        if y1 == 0 or y2 == 0:
            continue
        if y1 * y2 < 0:
            r = bisect_root_scalar(f, x[k], x[k+1])
            if r is not None:
                roots.append(float(r))

    roots = sorted(roots)
    dedup = []
    for r in roots:
        if not dedup or abs(r - dedup[-1]) > 1e-6:
            dedup.append(r)
    return [r for r in dedup if (a < r < b)]

def q_coeffs_edge(n, beta):
    coeffs = np.zeros(2*n + 1, dtype=np.float64)
    coeffs[0] = 1.0
    coeffs[1] = -1.0
    coeffs[n] = float(beta)
    coeffs[-2] = -1.0
    coeffs[-1] = 1.0
    return coeffs

def q_roots_edge(n, beta):
    return np.roots(q_coeffs_edge(n, beta))

def count_unimodular_roots(roots, unit_tol=UNIT_TOL):
    mods = np.abs(roots)
    return int(np.sum(np.abs(mods - 1.0) < unit_tol))

# -----------------------------
# Integrated scan
# -----------------------------
detail_rows = []
global_rows = []

for n in range(N_MIN, N_MAX + 1):
    for beta in BETAS:
        kappas = []
        all_unique_F = True
        all_Fprime_neg = True
        all_endpoint_sign_ok = True
        all_local_rule_ok = True

        for j in range(0, 2*n - 1):
            a, b = partition_interval(n, j)
            width = b - a
            eps = EDGE_EPS_FACTOR * width

            # Week 7 / uniqueness sanity
            x = np.linspace(a + eps, b - eps, GRID_PER_INTERVAL)
            fp_vals = Fprime(n, x)
            all_Fprime_negative_interval = bool(np.all(fp_vals < 0))
            F_roots = locate_roots_of_F_in_interval(n, j)
            num_F_roots = len(F_roots)
            F_left = F(n, a + eps)
            F_right = F(n, b - eps)
            endpoint_sign_ok = (F_left > 0) and (F_right < 0)

            # Week 5 / local rule
            sign_expected = interval_sign_expected(j)
            sign_matches_beta = (sign_expected == beta)

            if num_F_roots == 1:
                crit = F_roots[0]
                M_nj = abs(H(n, crit))
            else:
                M_nj = float(np.max(np.abs(H(n, x))))

            level_roots = locate_roots_of_level_in_interval(n, j, beta)
            actual_count = len(level_roots)

            if not sign_matches_beta:
                predicted_count = 0
                regime = "sign-mismatch"
            else:
                if M_nj < 1.0 - LEVEL_TOL:
                    predicted_count = 0
                    regime = "below"
                elif M_nj > 1.0 + LEVEL_TOL:
                    predicted_count = 2
                    regime = "above"
                else:
                    predicted_count = 1
                    regime = "tangent"

            local_rule_ok = (predicted_count == actual_count)

            kappa = actual_count
            kappas.append(kappa)

            all_unique_F &= (num_F_roots == 1)
            all_Fprime_neg &= all_Fprime_negative_interval
            all_endpoint_sign_ok &= endpoint_sign_ok
            all_local_rule_ok &= local_rule_ok

            detail_rows.append({
                "n": n,
                "beta": beta,
                "j": j,
                "num_F_roots": num_F_roots,
                "all_Fprime_negative_interval": all_Fprime_negative_interval,
                "endpoint_sign_ok": endpoint_sign_ok,
                "sign_matches_beta": sign_matches_beta,
                "M_nj": M_nj,
                "regime": regime,
                "predicted_count": predicted_count,
                "actual_count": actual_count,
                "local_rule_ok": local_rule_ok,
                "kappa_nj": kappa,
            })

        U_interval_sum = int(np.sum(kappas))
        U_poly = count_unimodular_roots(q_roots_edge(n, beta))

        global_rows.append({
            "n": n,
            "beta": beta,
            "week7_unique_F": all_unique_F,
            "week7_Fprime_negative": all_Fprime_neg,
            "week7_endpoint_sign_ok": all_endpoint_sign_ok,
            "week5_local_rule_ok": all_local_rule_ok,
            "week6_global_match": (U_poly == U_interval_sum),
            "U_polynomial": U_poly,
            "U_interval_sum": U_interval_sum,
            "difference": U_poly - U_interval_sum,
            "num_kappa_0": int(np.sum(np.array(kappas) == 0)),
            "num_kappa_1": int(np.sum(np.array(kappas) == 1)),
            "num_kappa_2": int(np.sum(np.array(kappas) == 2)),
            "all_pipeline_ok": (
                all_unique_F and
                all_Fprime_neg and
                all_endpoint_sign_ok and
                all_local_rule_ok and
                (U_poly == U_interval_sum)
            )
        })

detail_df = pd.DataFrame(detail_rows)
global_df = pd.DataFrame(global_rows)

fail_df = global_df[~global_df["all_pipeline_ok"]].copy()

# -----------------------------
# Print summary
# -----------------------------
print("="*78)
print("WEEK 7 CLOSEOUT — INTEGRATED SUMMARY OF WEEKS 4–7")
print("="*78)
print(f"n range scanned: {N_MIN} to {N_MAX}")
print(f"beta values scanned: {BETAS}")
print(f"Total (n,beta) cases checked: {len(global_df)}")
print(f"Total integrated failures: {len(fail_df)}")

retro_table(global_df, "Integrated summary by (n,beta)", max_rows=40)

if len(fail_df) == 0:
    print("\nNo integrated failures detected across the Week 4–7 pipeline.")
else:
    retro_table(
        fail_df[[
            "n", "beta",
            "week7_unique_F", "week7_Fprime_negative", "week7_endpoint_sign_ok",
            "week5_local_rule_ok", "week6_global_match",
            "U_polynomial", "U_interval_sum", "difference"
        ]],
        "Integrated failures",
        max_rows=30
    )

# -----------------------------
# Plot 1: pipeline success map
# -----------------------------
if MAKE_PLOTS:
    plot_df = global_df.copy()
    plot_df["all_pipeline_ok_int"] = plot_df["all_pipeline_ok"].astype(int)
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig1 = px.scatter(
        plot_df,
        x="n",
        y="U_polynomial",
        color="all_pipeline_ok_int",
        symbol="beta_str",
        hover_data=[
            "week7_unique_F", "week5_local_rule_ok",
            "week6_global_match", "difference"
        ],
        title="Weeks 4–7 integrated mechanism: pipeline check by (n,beta)"
    )
    apply_retro(fig1)
    fig1.update_layout(xaxis_title="n", yaxis_title="U(Q)")
    fig1.show()

# -----------------------------
# Plot 2: polynomial count vs interval sum
# -----------------------------
if MAKE_PLOTS:
    plot_df = global_df.copy()
    plot_df["beta_str"] = plot_df["beta"].astype(str)

    fig2 = px.scatter(
        plot_df,
        x="U_polynomial",
        y="U_interval_sum",
        color="beta_str",
        hover_data=["n", "difference", "num_kappa_0", "num_kappa_1", "num_kappa_2"],
        title="Integrated closeout: polynomial count vs interval contribution sum"
    )
    apply_retro(fig2)
    minv = min(plot_df["U_polynomial"].min(), plot_df["U_interval_sum"].min())
    maxv = max(plot_df["U_polynomial"].max(), plot_df["U_interval_sum"].max())
    fig2.add_trace(go.Scatter(
        x=[minv, maxv],
        y=[minv, maxv],
        mode="lines",
        name="y=x",
        line=dict(width=1.5, dash="dash", color=NEON_GREEN)
    ))
    fig2.update_layout(xaxis_title="U_polynomial", yaxis_title="U_interval_sum")
    fig2.show()

# -----------------------------
# Plot 3: diagnostic F_n profile for one n
# -----------------------------
if MAKE_PLOTS and (N_PLOT >= N_MIN) and (N_PLOT <= N_MAX):
    n = N_PLOT
    fig3 = go.Figure()

    for j in range(0, 2*n - 1):
        a, b = partition_interval(n, j)
        width = b - a
        eps = EDGE_EPS_FACTOR * width
        x = np.linspace(a + eps, b - eps, 500)
        y = F(n, x)

        fig3.add_trace(go.Scatter(
            x=x, y=y,
            mode="lines",
            showlegend=False,
            line=dict(width=1.2)
        ))

    fig3.add_trace(go.Scatter(
        x=[0, 2*np.pi], y=[0, 0],
        mode="lines",
        name="y=0",
        line=dict(width=1.5, dash="dash", color=NEON_GREEN)
    ))

    root_x = []
    root_y = []
    for j in range(0, 2*n - 1):
        roots = locate_roots_of_F_in_interval(n, j)
        if len(roots) == 1:
            root_x.append(roots[0])
            root_y.append(0.0)

    fig3.add_trace(go.Scatter(
        x=root_x, y=root_y,
        mode="markers",
        name="unique roots of F_n",
        marker=dict(size=6, color=NEON_PINK)
    ))

    apply_retro(fig3, f"Week 7 closeout diagnostic for F_{n}(t)")
    fig3.update_layout(xaxis_title="t", yaxis_title="F_n(t)")
    fig3.show()

# -----------------------------
# CSV export
# -----------------------------
if SAVE_CSV:
    global_path = f"{OUT_DIR}/week7_closeout_global_summary.csv"
    detail_path = f"{OUT_DIR}/week7_closeout_detail.csv"
    fail_path = f"{OUT_DIR}/week7_closeout_failures.csv"

    global_df.to_csv(global_path, index=False)
    detail_df.to_csv(detail_path, index=False)
    fail_df.to_csv(fail_path, index=False)

    print("\nSaved CSV files:")
    print(global_path)
    print(detail_path)
    print(fail_path)

# -----------------------------
# Suggested manuscript / note sentence
# -----------------------------
if len(fail_df) == 0:
    print(
        f"\nSuggested closeout statement:\n"
        f'Across the tested range n in [{N_MIN}, {N_MAX}] and beta=±1, the full interval mechanism '
        f'was internally consistent: the auxiliary function F_n was strictly decreasing on each '
        f'partition interval with exactly one zero, the local 0/1/2 counting rule was exact, and '
        f'the resulting global interval sum matched the polynomial unimodular-zero count in every case.'
    )
else:
    print(
        "\nAt least one integrated failure was detected. Inspect fail_df before writing the weekly closeout."
    )
```
